### 4.2 Python as a Calculator — Arithmetic Operations

````python
# Basic Arithmetic Operations
print("Addition:       5 + 3  =", 5 + 3)
print("Subtraction:    10 - 4 =", 10 - 4)
print("Multiplication: 6 * 7  =", 6 * 7)
print("Division:       15 / 4 =", 15 / 4)      # Returns float
print("Floor Division: 15 // 4 =", 15 // 4)    # Returns integer (rounded down)
print("Modulus:        15 % 4 =", 15 % 4)       # Remainder
print("Exponent:       2 ** 10 =", 2 ** 10)     # Power

# In Jupyter, the last expression's result is automatically displayed
2 + 2
````

### 4.3 String Operations

````python
# Strings in Python
greeting = "Hello"
name = "Data Scientist"

# Concatenation
print(greeting + ", " + name + "!")

# f-strings (formatted strings) — the modern way!
print(f"{greeting}, {name}!")

# String methods
message = "  python is awesome  "
print(f"Original:    '{message}'")
print(f"Strip:       '{message.strip()}'")
print(f"Upper:       '{message.strip().upper()}'")
print(f"Title:       '{message.strip().title()}'")
print(f"Replace:     '{message.strip().replace('awesome', 'amazing')}'")

# String length
print(f"\nLength of '{name}': {len(name)} characters")

# String indexing & slicing
text = "Python"
print(f"\nFirst char: {text[0]}")
print(f"Last char:  {text[-1]}")
print(f"Slice [0:3]: {text[0:3]}")
print(f"Reverse:    {text[::-1]}")
````

## 5. Variables and Data Types

Python is **dynamically typed** — you don't need to declare the type of a variable. Python figures it out automatically.

> 💡 **Key Point:** Variables persist across cells in a Jupyter notebook!

````python
# Integer
age = 25
print(f"age = {age}, type: {type(age)}")

# Float
salary = 75000.50
print(f"salary = {salary}, type: {type(salary)}")

# String
city = "Bangalore"
print(f"city = '{city}', type: {type(city)}")

# Boolean
is_student = True
print(f"is_student = {is_student}, type: {type(is_student)}")

# List (ordered, mutable collection)
scores = [85, 92, 78, 95, 88]
print(f"scores = {scores}, type: {type(scores)}")

# Dictionary (key-value pairs)
student = {
    "name": "Alice",
    "age": 22,
    "course": "Data Science",
    "gpa": 3.8
}
print(f"student = {student}, type: {type(student)}")
````

````python
# Variables persist across cells!
# We can still access 'scores' and 'student' from the previous cell

print(f"Scores from previous cell: {scores}")
print(f"Student name: {student['name']}")
print(f"Student GPA: {student['gpa']}")

# Type conversion
x = "42"
print(f"\nx = '{x}' (type: {type(x)})")

x_int = int(x)
print(f"int(x) = {x_int} (type: {type(x_int)})")

x_float = float(x)
print(f"float(x) = {x_float} (type: {type(x_float)})")
````

## 7. Displaying Rich Output — Charts & Tables

One of the biggest advantages of Jupyter notebooks: **inline visualizations!**

````python
# A simple line chart with Matplotlib
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [15000, 18000, 22000, 19500, 25000, 28000]

plt.figure(figsize=(8, 4))
plt.plot(months, sales, marker='o', color='steelblue', linewidth=2, markersize=8)
plt.fill_between(range(len(months)), sales, alpha=0.1, color='steelblue')
plt.title('Monthly Sales Performance', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("👆 This chart was rendered directly inside the notebook!")
````

## 8. A Data Analysis Teaser — Pure Python

Before we learn Pandas and NumPy, let's see what Python can do with just **built-in features**:

````python
# Quick Data Analysis with built-in Python
# Imagine these are daily temperatures (°C) for a week
temperatures = [28.5, 31.2, 27.8, 33.1, 29.4, 35.0, 30.6]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

# Basic statistics using built-in functions
total = sum(temperatures)
count = len(temperatures)
average = total / count
minimum = min(temperatures)
maximum = max(temperatures)

print("=" * 45)
print("  📊 Weekly Temperature Report")
print("=" * 45)
for day, temp in zip(days, temperatures):
    bar = "█" * int(temp)
    print(f"  {day}: {temp:5.1f}°C  {bar}")
print("-" * 45)
print(f"  Average: {average:.1f}°C")
print(f"  Min:     {minimum}°C ({days[temperatures.index(minimum)]})")
print(f"  Max:     {maximum}°C ({days[temperatures.index(maximum)]})")
print(f"  Range:   {maximum - minimum:.1f}°C")
print("=" * 45)

# Preview: This same analysis takes 2 lines with Pandas!
# We'll learn that in Session 5 & 6
````
