## 1. Advanced NumPy Indexing and Views

NumPy indexing is powerful. You can use boolean masks, integer arrays, and fancy selection patterns.

### Indexing modes
- **Basic indexing:** scalar or slice selection
- **Fancy indexing:** sequences of indices
- **Boolean indexing:** mask arrays of True/False

### Views vs copies
- `arr[1:4]` returns a **view**
- `arr.copy()` returns an independent **copy**

> ⚠️ Special case — modifying a slice may also modify the original array. Use `copy()` if you need independent data.

````python
import numpy as np

arr = np.arange(10)
slice_view = arr[2:6]
copy_arr = arr[2:6].copy()
slice_view[0] = 99
copy_arr[1] = 88

print('original:', arr)
print('slice_view:', slice_view)
print('copy_arr:', copy_arr)
````

### Fancy indexing
Fancy indexing uses integer arrays or lists to select specific positions.

```python
arr = np.array([10, 20, 30, 40, 50])
index = [4, 1, 3]
arr[index]  # [50, 20, 40]
```

Fancy indexing always returns a copy, not a view.

````python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
index = [4, 1, 3]
selected = arr[index]
selected[0] = 999

print('arr:', arr)
print('selected:', selected)
````

### Boolean indexing
Select values by conditions.

```python
mask = (arr % 20 == 0)
arr[mask]
```

Boolean masks are especially useful for filtering data based on rules.

````python
import numpy as np
arr = np.arange(1, 11)
mask = (arr % 3 == 0)
print('mask:', mask)
print('filtered:', arr[mask])
````

## 2. Advanced Slicing and Reshaping

### Multi-dimensional slicing
```python
matrix = np.arange(16).reshape(4, 4)
matrix[1:3, 2:4]
```

### Step slices
```python
matrix[::2, ::2]
```

### Reshape rules
- `reshape()` returns a view when possible
- `reshape(-1, 1)` infers one dimension
- `np.ravel()` returns a view if possible

> ⚠️ Special case — `reshape` may fail if the array is not contiguous in memory. Use `arr.copy()` or `np.ascontiguousarray(arr)`.

````python
import numpy as np
matrix = np.arange(16).reshape(4, 4)
print('matrix:\n', matrix)
print('submatrix:\n', matrix[1:3, 2:4])
print('step slice:\n', matrix[::2, ::2])
print('reshaped:', matrix.reshape(2, 8))
````

## 3. Broadcasting Deep Dive

Broadcasting extends arrays with smaller dimensions to match larger ones.

### Broadening rules
1. If necessary, prepend dimensions of 1 to the smaller array
2. Dimensions must be equal or one of them must be 1

### Examples
- `(3, 1) + (1, 4) -> (3, 4)`
- `(2, 3) + (3,) -> (2, 3)`

> ⚠️ Special case — `(2, 3) + (2,)` fails because the last dimensions 3 and 2 are incompatible.

````python
import numpy as np
col = np.array([[1], [2], [3]])
row = np.array([10, 20, 30])
print(col + row)

matrix = np.array([[1, 2, 3], [4, 5, 6]])
vector = np.array([1, 2, 3])
print(matrix + vector)
````

### Broadcasting for preprocessing
```python
X = np.random.rand(5, 3)
mean = X.mean(axis=0)
X_centered = X - mean
```

This is exactly how feature centering is implemented in machine learning.

## 4. NumPy Utility Functions

### Concatenation and stacking
| Function | Use | Result |
|---|---|---|
| `np.concatenate` | join along existing axis | flexible |
| `np.stack` | combine arrays along new axis | adds dimension |
| `np.vstack`, `np.hstack` | vertical/horizontal stacking | convenient |

### Repetition and tiling
- `np.repeat(arr, repeats, axis=0)`
- `np.tile(arr, reps)`

### Conditionals and selection
- `np.where(condition, x, y)`
- `np.unique(arr)`
- `np.nonzero(mask)`

> **Data science note:** These utilities are common when engineering features from raw arrays.

````python
import numpy as np

arrays = [np.arange(3), np.arange(3, 6)]
print('concatenate:', np.concatenate(arrays))
print('stack:', np.stack(arrays, axis=0))
print('vstack:\n', np.vstack(arrays))
print('hstack:', np.hstack(arrays))

arr = np.array([1, 2, 2, 3])
print('repeat:', np.repeat(arr, 2))
print('tile:', np.tile(arr, 2))
print('where:', np.where(arr > 1, arr, -1))
print('unique:', np.unique(arr))
````

### Linear algebra convenience functions
- `np.linalg.norm`
- `np.dot`, `@`
- `np.linalg.pinv`
- `np.linalg.eig`

Use these functions when working with covariance matrices or PCA-ready data.

````python
import numpy as np
A = np.array([[2, 0], [0, 3]])
print('norm:', np.linalg.norm(A))
print('pinv:\n', np.linalg.pinv(A))
print('eigs:', np.linalg.eig(A))
````

## 5. NumPy Performance and Memory Layout

### Memory layout
- `C` order: row-major
- `F` order: column-major

### Important functions
- `np.ascontiguousarray()`
- `np.asfortranarray()`
- `arr.flags` to inspect memory layouts

> ⚠️ Special case — `arr.T` on a non-contiguous array may return a view with different stride patterns. That is usually fine, but some functions require contiguous memory.

````python
import numpy as np
arr = np.arange(12).reshape(3, 4)
transposed = arr.T
print('arr.flags.contiguous:', arr.flags['C_CONTIGUOUS'])
print('transposed flags:', transposed.flags['C_CONTIGUOUS'])
print('contiguous transposed:\n', np.ascontiguousarray(transposed).flags)
````

## 6. Introduction to Pandas

Pandas provides two core objects:
- `Series`: one-dimensional labeled array
- `DataFrame`: two-dimensional labeled table

| Feature | NumPy | Pandas |
|---|---|---|
| Data shape | `ndarray` | `Series` / `DataFrame` |
| Labels | implicit integer positions | explicit row/column labels |
| Missing values | awkward | built-in support |
| Tabular data | manual handling | native |
| Group operations | manual | `groupby` |

````python
import pandas as pd
import numpy as np

# Create Series and DataFrame
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Cathy'],
    'age': [25, 30, 22],
    'score': [88.5, 92.3, 79.0]
})

print('Series:')
print(s)
print('\nDataFrame:')
print(df)
````

### Creating DataFrames
- From dict of lists
- From list of dicts
- From NumPy arrays
- From CSV files (later sessions)

Pandas preserves labels and makes row/column access easy.

````python
import pandas as pd
import numpy as np

arr = np.arange(12).reshape(4, 3)
df_from_array = pd.DataFrame(arr, columns=['A', 'B', 'C'])
print(df_from_array)
````

## 7. Selection with `.loc` and `.iloc`

- `.loc[row_label, column_label]` uses labels
- `.iloc[row_index, column_index]` uses integer positions

> ⚠️ Special case — `.loc` includes the end label for slices, while `.iloc` excludes the end index, just like standard Python slicing.

````python
import pandas as pd

df = pd.DataFrame({
    'city': ['NYC', 'LA', 'Chicago', 'Houston'],
    'temperature': [72, 85, 60, 90],
    'humidity': [60, 55, 70, 65]
}, index=['a', 'b', 'c', 'd'])

print('loc rows a-c:')
print(df.loc['a':'c', ['city', 'temperature']])
print('\niloc rows 0-2:')
print(df.iloc[0:3, 0:2])
````

### Boolean filtering in Pandas

```python
df[df['temperature'] > 70]
```

This is the most common pattern for selecting rows that meet conditions.

````python
import pandas as pd

df = pd.DataFrame({
    'city': ['NYC', 'LA', 'Chicago', 'Houston'],
    'temperature': [72, 85, 60, 90],
    'humidity': [60, 55, 70, 65]
})

print(df[df['temperature'] > 70])
````

## 8. Basic DataFrame Operations

| Task | Example |
|---|---|
| Add a column | `df['score'] = df['A'] * 2` |
| Rename | `df.rename(columns={'A':'Alpha'})` |
| Drop | `df.drop(columns=['B'])` |
| Sort | `df.sort_values('score', ascending=False)` |
| Describe | `df.describe()` |

````python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

df['total'] = df['A'] + df['B']
print('added total column:')
print(df)

print('\nrenamed:')
print(df.rename(columns={'A':'alpha', 'B':'beta'}))

print('\ndropped B:')
print(df.drop(columns=['B']))

print('\nsorted by total:')
print(df.sort_values('total', ascending=False))
````

## 9. Missing Values and Cleaning

### Common functions
- `df.isna()` / `df.isnull()`
- `df.dropna()`
- `df.fillna(value)`
- `df.interpolate()`

> **Data science note:** Missing values are common in real datasets, and Pandas makes them easy to analyze and clean.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Cathy', 'David'],
    'age': [25, np.nan, 22, 28],
    'score': [88.5, 92.3, np.nan, 81.0]
})

print('isna:')
print(df.isna())
print('\ndropna:')
print(df.dropna())
print('\nfillna:')
print(df.fillna({'age': df['age'].mean(), 'score': df['score'].median()}))
````

## 10. GroupBy and Aggregation

`groupby` is central to analysis of grouped data.

Example workflow:
1. group rows by key
2. apply aggregation
3. reset index if needed

| Function | Meaning |
|---|---|
| `sum()` | add values |
| `mean()` | average |
| `count()` | row counts |
| `agg()` | multiple functions |

> ⚠️ Special case — `groupby` returns a grouped object. Call aggregation after grouping.

````python
import pandas as pd

df = pd.DataFrame({
    'team': ['A', 'A', 'B', 'B', 'C'],
    'points': [10, 12, 7, 8, 15],
    'assists': [5, 7, 4, 6, 9]
})

print(df.groupby('team').sum())
print('\nmean by team:')
print(df.groupby('team').mean())
print('\ncount by team:')
print(df.groupby('team').agg({'points': 'sum', 'assists': 'mean'}))
````

## 11. NumPy vs Pandas Comparison

| Dimension | NumPy | Pandas |
|---|---|---|
| Primary use | numerical arrays | tabular data analysis |
| Labels | position only | row/column labels |
| Missing data | poor support | native support |
| Data types | one dtype per array | mixed dtypes in one table |
| Syntax | array-based | table-based |

> **Data science relevance:** Use NumPy for fast numerical computations, and Pandas for data cleaning, transformation, and exploratory analysis.

## 12. Practical Workflow Example

1. Load data into Pandas
2. Clean missing values
3. Create features
4. Convert to NumPy arrays for modeling

This is the typical path from raw tabular data to machine learning input.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'date': pd.to_datetime(['2024-01-01', '2024-01-02', '2024-01-03']),
    'visitors': [120, 150, np.nan],
    'sales': [1000, 1100, 1150]
})

df_clean = df.copy()
df_clean['visitors'] = df_clean['visitors'].fillna(df_clean['visitors'].mean())
df_clean['sales_per_visitor'] = df_clean['sales'] / df_clean['visitors']

X = df_clean[['visitors', 'sales_per_visitor']].to_numpy()
y = df_clean['sales'].to_numpy()

print(df_clean)
print('\nNumPy X shape:', X.shape)
print('NumPy y shape:', y.shape)
````
