## 1. Data Types — In-Depth

Python has several built-in data types. Understanding them is **critical** before we move to data analysis.

### 1.1 Numeric Types: `int`, `float`, `complex`

**`int` (Integer)**
- Whole numbers with **no decimal point**: `42`, `-17`, `0`
- Python integers have **unlimited precision** — they can be as large as your memory allows (unlike C/Java where `int` overflows at ~2 billion)
- Use underscores for readability: `1_000_000` is the same as `1000000`

**`float` (Floating-Point)**
- Numbers with a **decimal point**: `3.14`, `-0.001`, `2.0`
- Stored using **IEEE 754 double-precision** (64-bit), giving ~15-17 significant digits
- ⚠️ **Special Case — Floating-Point Precision:** `0.1 + 0.2` does NOT equal `0.3` exactly! This is because decimal fractions like 0.1 cannot be represented exactly in binary. This matters in financial calculations and comparisons
- Supports **scientific notation**: `6.022e23` means $6.022 \times 10^{23}$

**`complex` (Complex Numbers)**
- Numbers with a **real and imaginary part**: `3 + 4j`
- The imaginary unit is `j` (not `i` like in math)
- Used in signal processing, electrical engineering, and some ML algorithms (e.g., Fourier transforms)

````python
# --- Integer (int) ---
# Integers have unlimited precision in Python
a = 42
b = -17
big_num = 1_000_000_000  # Underscores for readability (Python 3.6+)

print(f"a = {a}  → type: {type(a)}")
print(f"b = {b}  → type: {type(b)}")
print(f"big_num = {big_num}  → type: {type(big_num)}")

# Python can handle VERY large integers natively
huge = 2 ** 100
print(f"\n2^100 = {huge}")
print(f"Number of digits: {len(str(huge))}")

# --- Float (floating-point) ---
pi = 3.14159
temperature = -40.0
scientific = 6.022e23  # Scientific notation (Avogadro's number)

print(f"\npi = {pi}  → type: {type(pi)}")
print(f"temperature = {temperature}")
print(f"Avogadro's number = {scientific}")

# ⚠️ Floating-point precision gotcha!
print(f"\n0.1 + 0.2 = {0.1 + 0.2}")  # Not exactly 0.3!
print(f"0.1 + 0.2 == 0.3? {0.1 + 0.2 == 0.3}")

# --- Complex ---
z = 3 + 4j
print(f"\nz = {z}  → type: {type(z)}")
print(f"Real part: {z.real}, Imaginary part: {z.imag}")
````

### 1.2 Boolean Type: `bool`

Booleans are the foundation of all decision-making in programming.

- Only two values: `True` and `False` (capitalized!)
- ⚠️ **Special Case — Booleans are integers!** `True` is actually `1` and `False` is `0`. You can do arithmetic with them: `True + True = 2`
- This is useful in data science: `sum([True, False, True, True])` counts the `True` values (result: 3)

**Truthy and Falsy Values:**
Every Python value can be evaluated as a boolean. This is crucial because `if my_list:` checks whether the list is non-empty.

| Falsy (evaluates to `False`) | Truthy (evaluates to `True`) |
|------------------------------|------------------------------|
| `0`, `0.0`, `0j` | Any non-zero number |
| `""` (empty string) | Any non-empty string |
| `[]`, `()`, `{}`, `set()` | Any non-empty collection |
| `None` | Everything else |
| `False` | `True` |

> 💡 **Why this matters:** When cleaning data, you often write `if value:` to skip `None`, `""`, `0`, and empty lists — all in one check!

````python
# Booleans: True or False
is_active = True
is_admin = False

print(f"is_active = {is_active} → type: {type(is_active)}")
print(f"is_admin = {is_admin}")

# Booleans are actually integers underneath!
print(f"\nTrue + True = {True + True}")    # 2
print(f"True * 10 = {True * 10}")          # 10
print(f"False + 5 = {False + 5}")          # 5

# Truthy and Falsy values — VERY important in Python!
# These are FALSY (evaluate to False):
print("\n--- Falsy Values ---")
falsy_values = [0, 0.0, "", [], {}, (), None, False]
for val in falsy_values:
    print(f"  bool({str(val):10s}) → {bool(val)}")

# Everything else is TRUTHY
print("\n--- Truthy Values ---")
truthy_values = [1, -1, 3.14, "hello", [1, 2], {"a": 1}, (1,)]
for val in truthy_values:
    print(f"  bool({str(val):15s}) → {bool(val)}")
````

### 1.3 String Type: `str` — Deep Dive

Strings are **immutable sequences of characters** — once created, you can't change individual characters (you create a new string instead).

**Creating Strings:**
- **Single quotes** `'hello'` and **double quotes** `"hello"` are identical — use whichever is convenient
- **Triple quotes** `"""..."""` or `'''...'''` for multi-line strings and docstrings
- **Raw strings** `r"..."` treat backslashes as literal characters — essential for Windows file paths and regex patterns

**Escape Characters:**
| Escape | Meaning | Example |
|--------|---------|---------|
| `\n` | Newline | `"line1\nline2"` |
| `\t` | Tab | `"col1\tcol2"` |
| `\\` | Literal backslash | `"C:\\Users"` |
| `\"` | Quote inside quotes | `"He said \"hi\""` |

> ⚠️ **Special Case — Immutability:** `text[0] = "H"` will raise a `TypeError`! To "change" a string, use methods like `.replace()` which return a NEW string.

````python
# Creating strings — single, double, or triple quotes
single = 'Hello'
double = "World"
multi_line = """This is a
multi-line string.
Very useful for documentation!"""

print(single, double)
print(multi_line)

# Escape characters
print("\n--- Escape Characters ---")
print("Tab:\tHello")
print("Newline:\nNext line")
print("Backslash: \\")
print("Quote: \"quoted\"")

# Raw strings (ignore escape characters) — useful for file paths
path = r"C:\Users\new_folder\test"
print(f"\nRaw string path: {path}")
````

````python
# String methods — your data cleaning toolkit!
text = "  Hello, Python World!  "

print(f"Original:     '{text}'")
print(f"strip():      '{text.strip()}'")
print(f"lower():      '{text.strip().lower()}'")
print(f"upper():      '{text.strip().upper()}'")
print(f"title():      '{text.strip().title()}'")
print(f"replace():    '{text.strip().replace('World', 'Data Science')}'")
print(f"split():      {text.strip().split()}")           # Split into list
print(f"split(','):   {text.strip().split(',')}")        # Split by comma

# Joining strings
words = ["Python", "is", "awesome"]
sentence = " ".join(words)
print(f"\nJoin list:    '{sentence}'")

# String checks
email = "user@example.com"
print(f"\n'{email}'")
print(f"  startswith('@'):   {email.startswith('@')}")
print(f"  endswith('.com'):  {email.endswith('.com')}")
print(f"  '@' in email:     {'@' in email}")
print(f"  isdigit():        {email.isdigit()}")
print(f"  find('@'):         {email.find('@')}")   # Index of '@'
print(f"  count('.'):        {email.count('.')}")
````

**String Methods — Your Data Cleaning Toolkit:**

These methods are **non-destructive** — they return a new string and leave the original unchanged.

| Method | What it does | Data Science Use |
|--------|-------------|-----------------|
| `.strip()` | Remove leading/trailing whitespace | Clean CSV column values |
| `.lower()` / `.upper()` | Change case | Standardize categories |
| `.replace(old, new)` | Replace substrings | Fix typos, clean values |
| `.split(sep)` | Split into list | Parse delimited data |
| `sep.join(list)` | Join list into string | Reconstruct text |
| `.startswith()` / `.endswith()` | Check prefix/suffix | Filter file names, URLs |
| `.find(sub)` | Find index of substring (-1 if not found) | Locate patterns |
| `.count(sub)` | Count occurrences | Frequency analysis |
| `.isdigit()` | Check if all digits | Validate numeric strings |

> ⚠️ **Special Case — `split()` vs `split(' ')`:** `split()` with no arguments splits on ANY whitespace and removes empty strings. `split(' ')` splits only on spaces and keeps empty strings. For data cleaning, `split()` is usually what you want.

````python
# String formatting — 3 ways (f-strings are the best!)
name = "Alice"
score = 95.678

# Method 1: % formatting (old style)
print("Hello %s, your score is %.2f%%" % (name, score))

# Method 2: .format() 
print("Hello {}, your score is {:.2f}%".format(name, score))

# Method 3: f-strings (Python 3.6+) ✅ RECOMMENDED
print(f"Hello {name}, your score is {score:.2f}%")

# f-string with expressions
print(f"2 + 3 = {2 + 3}")
print(f"Name in uppercase: {name.upper()}")
print(f"Score rounded: {round(score)}")

# f-string alignment and padding
for item, price in [("Apple", 1.5), ("Banana", 0.75), ("Cherry", 3.25)]:
    print(f"  {item:<10s} ${price:>6.2f}")
````

**String Formatting — 3 Methods (f-strings are the modern standard):**

| Method | Syntax | When to Use |
|--------|--------|-------------|
| `%` formatting | `"Hello %s" % name` | Legacy code only (Python 2 era) |
| `.format()` | `"Hello {}".format(name)` | Older Python 3, template strings |
| **f-strings** ✅ | `f"Hello {name}"` | **Always use this!** (Python 3.6+) |

**f-string Format Specifiers:**

| Specifier | Meaning | Example | Output |
|-----------|---------|---------|--------|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `3.14` |
| `:,` | Thousands separator | `f"{1000000:,}"` | `1,000,000` |
| `:<10` | Left-align, width 10 | `f"{'hi':<10}"` | `'hi        '` |
| `:>10` | Right-align, width 10 | `f"{'hi':>10}"` | `'        hi'` |
| `:^10` | Center, width 10 | `f"{'hi':^10}"` | `'    hi    '` |
| `:.2%` | Percentage | `f"{0.856:.2%}"` | `85.60%` |

> ⚠️ **Special Case — f-strings can contain ANY valid Python expression:** `f"{len(name)}"`, `f"{x if x > 0 else 'N/A'}"`, even `f"{', '.join(my_list)}"`. But keep them readable!

### 1.4 `None` Type

`None` represents the **absence of a value**. It's Python's version of `null` / `nil` in other languages.

**Key facts about `None`:**
- There is only **ONE** `None` object in memory — all `None` variables point to the same object
- ⚠️ **Special Case — Always use `is None`, never `== None`:** Since there's only one `None` object, `is` checks identity (same object) while `==` checks equality (can be overridden). Using `is` is both faster and more correct
- `None` is returned by functions that don't explicitly return anything
- In data science, `None` represents **missing data** — Pandas converts it to `NaN` (Not a Number) in numeric columns

````python
# None — the absence of a value
result = None
print(f"result = {result} → type: {type(result)}")

# Check for None using 'is' (not ==)
if result == None:
    print("Result has no value yet")

# Common use: default function parameters, missing data
# In data science, None often represents missing values (NaN in Pandas)
````

### 1.5 Type Conversion (Casting)

Converting between types is essential when reading data from files, APIs, or user input — data often arrives as strings.

**Conversion Functions:**

| Function | What it does | Gotcha |
|----------|-------------|--------|
| `int(x)` | Convert to integer | ⚠️ **Truncates** floats (doesn't round!): `int(3.9)` → `3` |
| `float(x)` | Convert to float | Works on strings: `float("3.14")` → `3.14` |
| `str(x)` | Convert to string | Everything can become a string |
| `bool(x)` | Convert to boolean | See truthy/falsy table above |
| `list(x)` | Convert to list | Splits strings into characters: `list("abc")` → `['a', 'b', 'c']` |
| `round(x, n)` | Round to n decimals | ⚠️ **Banker's rounding**: `round(0.5)` → `0`, `round(1.5)` → `2` (rounds to nearest even!) |

> ⚠️ **Special Case — `int()` truncates, `round()` uses banker's rounding:**
> - `int(3.9)` → `3` (cuts off the decimal, always towards zero)
> - `round(3.5)` → `4` but `round(4.5)` → `4` (rounds to nearest **even** number!)
> - For normal rounding, use `math.floor()` or `math.ceil()` from the `math` module

````python
# Type Conversion examples
print("--- String → Number ---")
age_str = "25"
age_int = int(age_str)
age_float = float(age_str)
print(f"'{age_str}' → int: {age_int}, float: {age_float}")

print("\n--- Number → String ---")
price = 49.99
price_str = str(price)
print(f"{price} → '{price_str}'")

print("\n--- Float ↔ Int ---")
x = 3.7
print(f"int(3.7) = {int(x)}")        # Truncates (not rounds!)
print(f"round(3.7) = {round(x)}")    # Rounds properly
print(f"round(3.14159, 2) = {round(3.14159, 2)}")  # Round to 2 decimals

print("\n--- To Boolean ---")
print(f"bool(0) = {bool(0)}")
print(f"bool(1) = {bool(1)}")
print(f"bool('') = {bool('')}")
print(f"bool('hello') = {bool('hello')}")

print("\n--- To List ---")
print(f"list('Python') = {list('Python')}")
print(f"list(range(5)) = {list(range(5))}")
````

## 2. Operators

### 2.1 Arithmetic Operators

| Operator | Description | Example | Result |
|----------|-------------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division (always returns float) | `7 / 2` | `3.5` |
| `//` | Floor Division (rounds DOWN) | `7 // 2` | `3` |
| `%` | Modulus (remainder) | `7 % 2` | `1` |
| `**` | Exponent (power) | `2 ** 3` | `8` |

> ⚠️ **Special Case — `/` vs `//`:** Division `/` ALWAYS returns a float, even for whole results: `6 / 3` → `2.0`. Floor division `//` rounds DOWN (not towards zero): `-7 // 2` → `-4` (not `-3`!)

> ⚠️ **Special Case — `%` with negatives:** Python's modulus always returns a value with the same sign as the divisor: `-7 % 2` → `1` (not `-1` like in C/Java). This follows the mathematical definition.

**Assignment Operators (shorthand):**

| Operator | Equivalent | Example |
|----------|-----------|---------|
| `+=` | `x = x + 5` | `x += 5` |
| `-=` | `x = x - 5` | `x -= 5` |
| `*=` | `x = x * 5` | `x *= 5` |
| `/=` | `x = x / 5` | `x /= 5` |
| `//=` | `x = x // 5` | `x //= 5` |
| `**=` | `x = x ** 2` | `x **= 2` |

### 2.2 Comparison Operators

These return `True` or `False` — you'll use them constantly in `if` statements and data filtering.

| Operator | Description | Example | Result |
|----------|-------------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater or equal | `5 >= 5` | `True` |
| `<=` | Less or equal | `5 <= 3` | `False` |

> ⚠️ **Special Case — `==` vs `=`:** `=` is assignment (`x = 5`), `==` is comparison (`x == 5`). Mixing them up is one of the most common beginner bugs!

> ⚠️ **Special Case — Chained Comparisons:** Python uniquely allows `18 < age < 30` which is equivalent to `age > 18 and age < 30`. This is cleaner and more readable than most languages.

````python
# Comparison Operators in action
age = 20

print(f"age = {age}")
print(f"age == 20  → {age == 20}")
print(f"age != 18  → {age != 18}")
print(f"age > 18   → {age > 18}")
print(f"age < 25   → {age < 25}")
print(f"age >= 20  → {age >= 20}")
print(f"age <= 19  → {age <= 19}")

# Chained comparisons — unique Python feature!
print(f"\n18 < age < 30 → {18 < age < 30}")  # Between 18 and 30
print(f"1 < 2 < 3 < 4 → {1 < 2 < 3 < 4}")
````

### 2.3 Logical Operators: `and`, `or`, `not`

| Operator | Description | Example | Result |
|----------|-------------|---------|--------|
| `and` | Both must be True | `True and False` → `False` |
| `or` | At least one True | `True or False` → `True` |
| `not` | Inverts the value | `not True` → `False` |

> ⚠️ **Special Case — Short-Circuit Evaluation:** Python stops evaluating as soon as the result is determined:
> - `False and anything` → `False` (doesn't evaluate `anything`)
> - `True or anything` → `True` (doesn't evaluate `anything`)
> - This is useful: `if x != 0 and 10/x > 2:` — the division never executes if `x` is `0`!

> ⚠️ **Special Case — `and`/`or` return VALUES, not booleans:**
> - `"hello" and "world"` → `"world"` (returns last truthy value)
> - `"" or "default"` → `"default"` (returns first truthy value)
> - This is used as a default pattern: `name = user_input or "Anonymous"`

**Identity vs Equality:**

| Operator | Checks | Use When |
|----------|--------|----------|
| `==` | Same **value** | Comparing values: `[1,2] == [1,2]` → `True` |
| `is` | Same **object** in memory | Checking `None`: `x is None` |

> ⚠️ **Special Case — `is` with small integers:** Python caches integers from -5 to 256, so `a = 5; b = 5; a is b` → `True`. But `a = 500; b = 500; a is b` → `False`! Always use `==` for value comparison.

**Membership Operators: `in`, `not in`**

Check if a value exists in a sequence (list, string, dict keys). Very Pythonic — use instead of looping to search!

````python
# Logical Operators
age = 25
income = 60000

# Combining conditions
print(f"age = {age}, income = {income}")
print(f"age > 18 and income > 50000 → {age > 18 and income > 50000}")
print(f"age > 30 or income > 50000  → {age > 30 or income > 50000}")
print(f"not (age > 30)              → {not (age > 30)}")

# Real-world example: Loan eligibility
has_good_credit = True
is_employed = True
has_collateral = False

eligible = (has_good_credit and is_employed) or has_collateral
print(f"\nLoan eligible: {eligible}")

# Identity operators: is, is not
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"\na == b: {a == b}")    # True — same VALUE
print(f"a is b: {a is b}")    # False — different OBJECTS in memory
print(f"a is c: {a is c}")    # True — same object (c points to a)

# Membership operators: in, not in
fruits = ["apple", "banana", "cherry"]
print(f"\n'banana' in fruits: {'banana' in fruits}")
print(f"'mango' not in fruits: {'mango' not in fruits}")
````

## 3. Conditional Constructs: `if`, `elif`, `else`

Conditionals let your program **make decisions** based on data values.

Python uses **indentation** (4 spaces) to define code blocks — no curly braces `{}` like C/Java/JS!

```python
if condition:          # condition evaluates to True/False
    # code runs if condition is True
elif another_condition:  # optional, can have multiple
    # code runs if first is False, this is True
else:                    # optional, only one
    # code runs if ALL above are False
```

**Key Rules:**
1. **Indentation is mandatory** — Python uses it to define scope. Mixing tabs and spaces causes `IndentationError`
2. **Colon `:` is required** after `if`, `elif`, `else`
3. `elif` is short for "else if" — use as many as you need
4. `else` is optional and catches everything not matched above
5. Conditions can be any expression that evaluates to a truthy/falsy value

> ⚠️ **Special Case — Empty blocks:** Python requires at least one line in every block. Use `pass` as a placeholder: `if condition: pass`

````python
# Simple if-else
temperature = 20

if temperature > 30:
    print(f"🔥 It's hot! ({temperature}°C)")
else:
    print(f"😊 Pleasant weather ({temperature}°C)")
````

````python
# if-elif-else chain — Grading System
score = 10

if score >= 90:
    grade = "A+"
    remark = "Outstanding!"
elif score >= 80:
    grade = "A"
    remark = "Excellent!"
elif score >= 70:
    grade = "B"
    remark = "Good job!"
elif score >= 60:
    grade = "C"
    remark = "Satisfactory"
elif score >= 50:
    grade = "D"
    remark = "Needs improvement"
else:
    grade = "F"
    remark = "Failed"

print(f"Score: {score} → Grade: {grade} ({remark})")
````

````python
# Nested if statements
age = 20
has_license = False
has_insurance = True

if age >= 18:
    if has_license:
        if has_insurance:
            print("✅ You can drive!")
        else:
            print("❌ You need insurance first")
    else:
        print("❌ You need a driving license first")
else:
    print("❌ You must be 18 or older to drive")

# Ternary operator (one-liner if-else)
status = "Adult" if age >= 18 else "Minor"
print(f"\nAge {age} → {status}")

# Practical: Check if a number is even or odd
num = 42
parity = "Even" if num % 2 == 0 else "Odd"
print(f"{num} is {parity}")
````

**Nested `if` Statements & Ternary Operator:**

- **Nested if:** Placing `if` inside another `if`. Works but can become hard to read — prefer combining with `and`/`or` when possible
- **Ternary operator:** `value_if_true if condition else value_if_false` — a one-liner `if-else` for simple assignments
  
> ⚠️ **Special Case — Nested ternary:** You CAN chain them: `"A" if x > 90 else "B" if x > 80 else "C"` — but avoid this for readability. Use `if/elif/else` for 3+ branches.

````python
# match-case (Python 3.10+) — Structural Pattern Matching
http_status = _

match http_status:
    case 200:
        message = "OK — Success"
    case 301:
        message = "Moved Permanently"
    case 404:
        message = "Not Found"
    case 500:
        message = "Internal Server Error"
    case _:
        message = "Unknown status code"

print(f"HTTP {http_status}: {message}")
````

**`match-case` — Structural Pattern Matching (Python 3.10+):**

This is Python's version of `switch-case` from other languages, but far more powerful — it can match patterns, not just values.

- The `_` (underscore) is the **wildcard/default** case — catches everything not matched above
- Unlike C's `switch`, there is **no fall-through** — only the matched case executes
- ⚠️ **Special Case — Requires Python 3.10+:** If you're on an older version, use `if/elif/else` instead. Check with `import sys; print(sys.version)`

## 4. Loops: `for` and `while`

Loops let you repeat actions — essential for processing data row by row, column by column.

### 4.1 `for` Loop

The `for` loop iterates over any **iterable** (list, string, range, dict, tuple, set, file, etc.)

```python
for item in iterable:
    # do something with item
```

**`range()` — Generate Number Sequences:**

| Usage | Result | Meaning |
|-------|--------|---------|
| `range(5)` | `0, 1, 2, 3, 4` | Start from 0, stop BEFORE 5 |
| `range(2, 8)` | `2, 3, 4, 5, 6, 7` | Start at 2, stop BEFORE 8 |
| `range(0, 10, 2)` | `0, 2, 4, 6, 8` | Start at 0, stop BEFORE 10, step by 2 |
| `range(10, 0, -1)` | `10, 9, 8, ..., 1` | Countdown (negative step) |

> ⚠️ **Special Case — `range()` excludes the stop value:** `range(5)` gives `0,1,2,3,4` (NOT 5). This is a common source of off-by-one errors. The "stop before" convention matches Python's slice notation `[start:stop]`.

> ⚠️ **Special Case — `range()` is lazy:** It doesn't create a list in memory. `range(1_000_000_000)` uses almost no memory! Wrap in `list()` only when you need the actual list.

````python
# for loop — iterating over a list
fruits = ["apple", "banana", "cherry", "dragonfruit", "elderberry"]

print("--- Iterating over a list ---")
for fruit in fruits:
    print(f"  I like {fruit}!")

# for loop with range()
print("\n--- range() examples ---")
print("range(5):", list(range(5)))          # 0, 1, 2, 3, 4
print("range(2, 8):", list(range(2, 8)))    # 2, 3, 4, 5, 6, 7
print("range(0, 10, 2):", list(range(0, 10, 2)))  # 0, 2, 4, 6, 8 (step=2)
print("range(10, 0, -2):", list(range(10, 0, -2)))  # Countdown

# Multiplication table
print("\n--- Multiplication Table for 7 ---")
for i in range(1, 11):
    print(f"  7 × {i:2d} = {7 * i:3d}")
````

````python
# enumerate() — get index AND value
print("--- enumerate() ---")
cities = ["Mumbai", "Delhi", "Bangalore", "Chennai"]
for index, city in enumerate(cities):
    print(f"  {index}: {city}")

# Starting from 1
print("\n--- enumerate(start=1) ---")
for rank, city in enumerate(cities, start=1):
    print(f"  #{rank}: {city}")

# zip() — iterate over multiple lists simultaneously
print("\n--- zip() ---")
students = ["Alice", "Bob", "Charlie"]
grades = [92, 85, 78]
subjects = ["Math", "Science", "English"]

for student, grade, subject in zip(students, grades, subjects):
    print(f"  {student}: {grade} in {subject}")

# Iterating over a dictionary
print("\n--- Dictionary iteration ---")
student_scores = {"Alice": 92, "Bob": 85, "Charlie": 78, "Diana": 95}

for name, score in student_scores.items():
    status = "PASS ✅" if score >= 80 else "FAIL ❌"
    print(f"  {name:10s}: {score} → {status}")
````

**Essential Loop Helpers: `enumerate()` and `zip()`**

| Function | What it does | When to use |
|----------|-------------|-------------|
| `enumerate(iterable, start=0)` | Gives you `(index, value)` pairs | When you need both the position AND the value |
| `zip(iter1, iter2, ...)` | Pairs up items from multiple iterables | When iterating over parallel lists together |

> ⚠️ **Special Case — `zip()` stops at the shortest iterable:** If lists have different lengths, `zip()` silently drops extras from the longer list. Use `itertools.zip_longest()` if you need all values.

> ⚠️ **Special Case — Don't use `for i in range(len(list))`!** This is a common beginner anti-pattern. Use `enumerate()` instead — it's more Pythonic, readable, and less error-prone.

**Dictionary Iteration:**
- `dict.keys()` — iterate over keys (default behavior)
- `dict.values()` — iterate over values  
- `dict.items()` — iterate over `(key, value)` pairs ← most useful!

### 4.2 `while` Loop

A `while` loop runs as long as the condition is `True`. Use it when you **don't know in advance** how many iterations you need.

```python
while condition:
    # body — runs repeatedly
    # MUST change condition eventually, or you get an infinite loop!
```

> ⚠️ **Special Case — Infinite Loops:** If the condition is always `True`, the loop runs forever! Always ensure something inside the loop changes the condition. In Jupyter, use the **Stop/Interrupt** button (⬛) if this happens.

> 💡 **`for` vs `while`:** Use `for` when iterating over a known sequence (90% of the time in data science). Use `while` for "keep going until a condition is met" (e.g., convergence in algorithms, user input validation).

````python
# while loop — countdown
print("--- Countdown ---")
count = 5
while count > 0:
    print(f"  {count}...")
    count -= 1
print("  🚀 Liftoff!")

# while with user-like simulation (no actual input in notebook)
print("\n--- Password checker simulation ---")
passwords = ["abc", "hello", "python123"]  # Simulating user attempts
correct = "python123"

for attempt in passwords:
    if attempt == correct:
        print(f"  Attempt '{attempt}' → ✅ Access granted!")
        break
    else:
        print(f"  Attempt '{attempt}' → ❌ Wrong password")
````

### 4.3 Loop Control: `break`, `continue`, `pass`

| Statement | What it does | When to use |
|-----------|-------------|-------------|
| `break` | Exit the loop **immediately** | Found what you're looking for |
| `continue` | Skip the **rest of this iteration**, go to next | Skip invalid/unwanted data |
| `pass` | Do **nothing** (placeholder) | Empty blocks that you'll fill later |

> ⚠️ **Special Case — `for-else`:** Python has a unique `for...else` construct. The `else` block runs **only if the loop completed without hitting `break`**. Useful for "search and report if not found" patterns.

> ⚠️ **Special Case — `break` only exits the innermost loop:** In nested loops, `break` only exits the loop it's in, not outer loops. To break out of multiple loops, use a flag variable or put the loops in a function and use `return`.

````python
# break — stop loop when condition is met
print("--- break: Find first number divisible by 7 ---")
for num in range(1, 100):
    if num % 7 == 0:
        print(f"  Found it: {num}")
        break

# continue — skip certain iterations
print("\n--- continue: Print only odd numbers 1-10 ---")
for num in range(1, 11):
    if num % 2 == 0:
        continue  # Skip even numbers
    print(f"  {num}", end=" ")
print()  # New line

# pass — placeholder (does nothing)
print("\n--- pass: Placeholder for future code ---")
for num in range(5):
    if num == 3:
        pass  # TODO: handle special case later
    else:
        print(f"  Processing {num}")

# for-else (runs if loop completes WITHOUT break)
print("\n--- for-else: Check if a number is prime ---")
num = 17
for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
        print(f"  {num} is NOT prime (divisible by {i})")
        break
else:
    print(f"  {num} IS prime! ✅")
````

### 4.4 Comprehensions — Compact Loop Syntax

Comprehensions are a **Pythonic** way to create lists, dicts, and sets in a single line. You'll use these **constantly** in data science.

**Syntax:**

```python
# List comprehension
[expression for item in iterable if condition]

# Dict comprehension
{key_expr: value_expr for item in iterable if condition}

# Set comprehension
{expression for item in iterable if condition}
```

**How to read a comprehension:** Read it as an English sentence:
- `[x**2 for x in range(5) if x % 2 == 0]`
- "Give me **x squared** **for each x** in 0-4 **if x is even**"

> ⚠️ **Special Case — `if` vs `if-else` placement:**
> - **Filter** (if only): Goes at the END → `[x for x in data if x > 0]`
> - **Transform** (if-else): Goes BEFORE `for` → `["even" if x%2==0 else "odd" for x in data]`
> - Mixing up the order is a very common syntax error!

> ⚠️ **Special Case — Don't overuse comprehensions!** If your comprehension needs nested loops and multiple conditions, it becomes unreadable. Use a regular loop instead. Rule of thumb: if it doesn't fit in one readable line, use a loop.

> 💡 **Generator expression:** Use `()` instead of `[]` for a **generator** that computes lazily: `sum(x**2 for x in range(1000000))` — uses almost no memory!

````python
# --- List Comprehension ---
# Traditional loop
squares_loop = []
for x in range(1, 6):
    squares_loop.append(x ** 2)
print(f"Loop:          {squares_loop}")

# Comprehension (same result, 1 line!)
squares_comp = [x ** 2 for x in range(1, 6)]
print(f"Comprehension: {squares_comp}")

# With condition (filter)
even_squares = [x ** 2 for x in range(1, 11) if x % 2 == 0]
print(f"Even squares:  {even_squares}")

# With if-else (transform)
labels = ["even" if x % 2 == 0 else "odd" for x in range(1, 6)]
print(f"Labels:        {labels}")

# --- Dict Comprehension ---
names = ["Alice", "Bob", "Charlie"]
name_lengths = {name: len(name) for name in names}
print(f"\nName lengths:  {name_lengths}")

# Fahrenheit to Celsius conversion
fahrenheit = {"NYC": 77, "LA": 86, "Chicago": 68}
celsius = {city: round((temp - 32) * 5/9, 1) for city, temp in fahrenheit.items()}
print(f"Celsius:       {celsius}")

# --- Set Comprehension ---
words = ["hello", "world", "hello", "python", "world"]
unique_lengths = {len(w) for w in words}
print(f"\nUnique word lengths: {unique_lengths}")
````

## 5. Functions

Functions let you **organize** and **reuse** code. In data science you'll write functions to clean data, build features, and create pipelines.

### 5.1 Defining & Calling Functions

```python
def function_name(param1, param2="default"):
    """Docstring — describes what the function does."""
    # function body
    return result
```

**Anatomy of a function:**
1. **`def`** keyword — starts the definition
2. **Name** — should be descriptive and use `snake_case`
3. **Parameters** — inputs the function receives (can have defaults)
4. **Docstring** — triple-quoted description (shows up in `help()` and IDE tooltips)
5. **Body** — the work the function does (indented)
6. **`return`** — sends a value back to the caller

**Parameter Types:**
| Type | Example | Description |
|------|---------|-------------|
| Positional | `def f(x, y)` | Order matters, required |
| Default | `def f(x, y=10)` | Has a fallback value, optional |
| Keyword | `f(y=20, x=5)` | Named at call site, order doesn't matter |

> ⚠️ **Special Case — Default mutable arguments:** NEVER use a mutable default like `def f(items=[])`. The list is shared across all calls! Use `def f(items=None): items = items or []` instead.

> ⚠️ **Special Case — Functions without `return`:** If a function has no `return` statement, it returns `None` implicitly.

````python
# Basic function
def greet(name):
    """Greet a person by name."""
    return f"Hello, {name}! Welcome to Data Science."

# Calling the function
message = greet("Alice")
print(message)
print(greet("Bob"))

# Function with multiple parameters and default values
def calculate_bmi(weight_kg, height_m, round_to=1):
    """Calculate Body Mass Index."""
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, round_to)

# Using positional arguments
print(f"\nBMI: {calculate_bmi(70, 1.75)}")

# Using keyword arguments (order doesn't matter)
print(f"BMI: {calculate_bmi(height_m=1.80, weight_kg=85)}")

# Using default parameter
print(f"BMI (2 decimals): {calculate_bmi(70, 1.75, round_to=2)}")
````

### 5.2 Returning Multiple Values

Python functions can return **multiple values** as a **tuple** — a feature many languages lack. This is extremely useful for statistics functions that return mean, median, std dev, etc.

```python
def stats(data):
    return mean, median, std_dev  # Returns a tuple

m, md, sd = stats(data)  # Unpack into separate variables
```

> ⚠️ **Special Case — Tuple unpacking:** The number of variables on the left MUST match the number of returned values, or you get a `ValueError`. Use `_` for values you want to ignore: `mean, _, std = stats(data)`

````python
# Functions can return multiple values as a tuple
def analyze_scores(scores):
    """Return basic statistics for a list of scores."""
    avg = sum(scores) / len(scores)
    minimum = min(scores)
    maximum = max(scores)
    spread = maximum - minimum
    return avg, minimum, maximum, spread

test_scores = [85, 92, 78, 95, 88, 67, 91]

# Unpack the returned tuple
average, low, high, spread = analyze_scores(test_scores)
print(f"Scores:  {test_scores}")
print(f"Average: {average:.1f}")
print(f"Low:     {low}")
print(f"High:    {high}")
print(f"Spread:  {spread}")
````

### 5.3 `*args` and `**kwargs` — Flexible Arguments

These let functions accept a **variable number of arguments**:

| Syntax | Receives | Use Case |
|--------|----------|----------|
| `*args` | Tuple of positional arguments | `def mean(*numbers)` — pass any count of values |
| `**kwargs` | Dict of keyword arguments | `def profile(**details)` — pass any named fields |

**Parameter order rule:** `def func(regular, *args, **kwargs)` — regular params first, then `*args`, then `**kwargs`

> ⚠️ **Special Case — `*` and `**` for unpacking:** You can also USE `*` and `**` when CALLING a function:
> - `my_list = [1, 2, 3]; func(*my_list)` → same as `func(1, 2, 3)`
> - `my_dict = {"x": 1, "y": 2}; func(**my_dict)` → same as `func(x=1, y=2)`
> - This is called "argument unpacking" and you'll see it frequently in ML code

````python
# *args — accept any number of positional arguments
def calculate_mean(*numbers):
    """Calculate mean of any number of values."""
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

print(f"Mean(1, 2, 3):       {calculate_mean(1, 2, 3)}")
print(f"Mean(10, 20, 30, 40): {calculate_mean(10, 20, 30, 40)}")

# **kwargs — accept any number of keyword arguments
def create_profile(**details):
    """Create a user profile from keyword arguments."""
    print("\n📋 User Profile:")
    for key, value in details.items():
        print(f"  {key.replace('_', ' ').title()}: {value}")

create_profile(
    name="Alice",
    age=25,
    city="Bangalore",
    role="Data Scientist",
    experience_years=3
)

# Combining regular, *args, and **kwargs
def log_event(event_type, *tags, **metadata):
    """Log an event with tags and metadata."""
    print(f"\n📌 Event: {event_type}")
    print(f"   Tags: {', '.join(tags)}")
    for key, value in metadata.items():
        print(f"   {key}: {value}")

log_event("MODEL_TRAINED", "ml", "production", "v2",
          accuracy=0.95, model="RandomForest", dataset="loan_data")
````

### 5.4 Lambda Functions — Anonymous One-Liners

Lambda functions are small, **anonymous** (unnamed) functions defined in a single expression.

```python
lambda parameters: expression
```

- Can only contain a **single expression** (no statements, loops, or multi-line logic)
- Returns the result of the expression automatically (no `return` keyword)
- Primarily used as **throwaway functions** passed to `map()`, `filter()`, `sorted()`

**Comparison: `def` vs `lambda`**

```python
# These are equivalent:
def square(x):      |  square = lambda x: x ** 2
    return x ** 2   |
```

**Common patterns with lambda:**

| Pattern | What it does | Example |
|---------|-------------|---------|
| `map(func, iterable)` | Apply func to **every item** | `list(map(lambda x: x*2, [1,2,3]))` → `[2,4,6]` |
| `filter(func, iterable)` | Keep items where func returns **True** | `list(filter(lambda x: x>0, [-1,2,-3,4]))` → `[2,4]` |
| `sorted(iterable, key=func)` | Sort by custom **key** | `sorted(names, key=lambda s: len(s))` |

> ⚠️ **Special Case — Lambda vs comprehension:** In modern Python, **list comprehensions are preferred** over `map`/`filter` with lambda: `[x*2 for x in items]` is clearer than `list(map(lambda x: x*2, items))`. However, lambda is essential for `sorted(key=...)` and Pandas `.apply()`.

> ⚠️ **Special Case — Don't assign lambda to a variable!** `square = lambda x: x**2` technically works but PEP 8 says use `def` instead. Lambda is for inline, throwaway usage.

````python
# Lambda: anonymous function in one line
# Syntax: lambda parameters: expression

# Regular function
def square(x):
    return x ** 2

# Same thing as lambda
square_lambda = lambda x: x ** 2

print(f"square(5) = {square(5)}")
print(f"square_lambda(5) = {square_lambda(5)}")

# Lambda with map() — apply function to every item
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print(f"\nSquared: {squared}")

# Lambda with filter() — keep items that return True
scores = [45, 82, 67, 91, 55, 78, 33, 88]
passed = list(filter(lambda x: x >= 60, scores))
print(f"Passed (>=60): {passed}")

# Lambda with sorted() — custom sorting
students = [
    {"name": "Alice", "gpa": 3.8},
    {"name": "Bob", "gpa": 3.2},
    {"name": "Charlie", "gpa": 3.9},
    {"name": "Diana", "gpa": 3.5},
]

# Sort by GPA (descending)
sorted_students = sorted(students, key=lambda s: s["gpa"], reverse=True)
print("\nStudents by GPA (highest first):")
for s in sorted_students:
    print(f"  {s['name']}: {s['gpa']}")
````

## 6. Error Handling: `try`, `except`, `finally`, `raise`

Errors **will** happen — especially when loading messy real-world data. Good error handling makes your code **robust** instead of crashing.

**Python has two types of errors:**
1. **Syntax Errors** — Invalid code structure (caught before running). Fix: correct the code
2. **Exceptions** — Runtime errors (caught during execution). Fix: handle with `try/except`

### Common Exception Types

| Exception | When it happens | Example |
|-----------|----------------|---------|
| `SyntaxError` | Invalid Python syntax | `print("hello"` (missing `)`) |
| `NameError` | Variable not defined | `print(undefined_var)` |
| `TypeError` | Wrong data type for operation | `"hello" + 5` |
| `ValueError` | Right type but wrong value | `int("hello")` |
| `IndexError` | List index out of range | `[1,2,3][10]` |
| `KeyError` | Dictionary key doesn't exist | `{"a":1}["b"]` |
| `FileNotFoundError` | File path is wrong | `open("nonexistent.csv")` |
| `ZeroDivisionError` | Division by zero | `10 / 0` |
| `AttributeError` | Object doesn't have that method | `5.upper()` |

**The error handling structure:**
```python
try:
    # Code that MIGHT fail
except SpecificError as e:
    # Handle that specific error (can have multiple except blocks)
else:
    # Runs ONLY if NO exception occurred (optional)
finally:
    # ALWAYS runs, even if exception occurred (optional, used for cleanup)
```

> ⚠️ **Special Case — Catch specific exceptions, NEVER bare `except:`!** Using `except:` or `except Exception:` catches ALL errors, including typos and bugs you'd want to know about. Always catch the specific error you expect.

````python
# --- Basic try/except ---
# Rule: Wrap ONLY the line(s) that might fail

# Example 1: Catching a specific error
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")

# Example 2: Accessing the error message with 'as e'
try:
    number = int("hello")
except ValueError as e:
    print(f"Conversion failed: {e}")
    # Output: Conversion failed: invalid literal for int() with base 10: 'hello'

# Example 3: Multiple except blocks (order matters — most specific first!)
try:
    data = [1, 2, 3]
    value = data[10]
except IndexError:
    print("Index out of range!")
except TypeError:
    print("Wrong type used as index!")
except Exception as e:
    # Catch-all as LAST resort (logs unexpected errors)
    print(f"Unexpected error: {e}")

# ⚠️ Special Case: Bare 'except:' catches EVERYTHING, even KeyboardInterrupt!
# NEVER do this:
# try:
#     x = int("abc")
# except:          # ← BAD! Hides bugs and prevents Ctrl+C!
#     pass

# ✅ Instead, always specify the exception type:
try:
    x = int("abc")
except ValueError:
    print("Caught the specific error we expected")
````

### `try` / `except` / `else` / `finally` — Full Structure

**Execution flow:**
```
try → (error?) → YES → except → finally
                  NO  → else   → finally
```

| Block | Runs when | Use case |
|-------|----------|----------|
| `try` | Always (code that might fail) | Wrap risky operations |
| `except` | Only if `try` raised an error | Handle the error gracefully |
| `else` | Only if `try` succeeded (no errors) | Code that depends on `try` succeeding |
| `finally` | **ALWAYS** — even after `return`, `break`, or unhandled exceptions | Cleanup: close files, DB connections |

> ⚠️ **Special Case — `else` is NOT the same as putting code at the end of `try`!** Code in `else` runs only on success AND is NOT protected by `except`. This makes bugs in `else` visible instead of silently caught.

> ⚠️ **Special Case — `finally` runs even after `return`!** If both `try` and `finally` have `return`, the `finally` return wins (overrides).

````python
# --- try / except / else / finally in action ---

# Example 1: All four blocks
try:
    number = int("42")       # This succeeds
except ValueError:
    print("Not a valid number!")
else:
    # Runs because try succeeded
    print(f"Converted successfully: {number}")
finally:
    # ALWAYS runs
    print("Attempt complete.")
# Output: Converted successfully: 42
#         Attempt complete.

print("---")

# Example 2: When try fails
try:
    number = int("hello")    # This fails → jumps to except
except ValueError:
    print("Not a valid number!")
else:
    # Skipped because try failed
    print(f"Converted: {number}")
finally:
    # Still runs!
    print("Attempt complete.")
# Output: Not a valid number!
#         Attempt complete.

print("---")

# Example 3: Data Science use case — safely loading a file
import os

try:
    filename = "data.csv"
    # Simulate checking if file exists
    if not os.path.exists(filename):
        raise FileNotFoundError(f"'{filename}' not found")
except FileNotFoundError as e:
    print(f"Error: {e}")
    print("Using default data instead...")
else:
    print("File loaded successfully!")
finally:
    print("Data loading attempt finished.")

print("---")

# ⚠️ Special Case: finally overrides return!
def risky_function():
    try:
        return "from try"
    finally:
        return "from finally"    # ← This wins! (bad practice, avoid)

print(f"Return value: {risky_function()}")
# Output: Return value: from finally
````

### `raise` — Throwing Your Own Errors

Use `raise` to **deliberately** create an error when something is logically wrong (even if Python wouldn't crash).

```python
raise ErrorType("Your message")
```

**When to use `raise`:**
- Validating function inputs (e.g., negative age, empty dataset)
- Enforcing business rules (e.g., price can't be negative)
- Converting a warning into a hard stop

**Exception Hierarchy (simplified):**
```
BaseException
 ├── KeyboardInterrupt    (Ctrl+C — don't catch this!)
 ├── SystemExit           (sys.exit() — don't catch this!)
 └── Exception            (all "normal" errors inherit from here)
      ├── ValueError
      ├── TypeError
      ├── KeyError
      ├── IndexError
      ├── FileNotFoundError
      └── ... (hundreds more)
```

> ⚠️ **Special Case — Creating custom exceptions**: For larger projects, create your own exception classes by inheriting from `Exception`. This lets callers handle YOUR specific errors differently from generic ones.

````python
# --- raise: Throwing your own errors ---

# Example 1: Validating function input
def calculate_bmi(weight_kg, height_m):
    if weight_kg <= 0:
        raise ValueError(f"Weight must be positive, got {weight_kg}")
    if height_m <= 0:
        raise ValueError(f"Height must be positive, got {height_m}")
    return weight_kg / (height_m ** 2)

# Valid call
print(f"BMI: {calculate_bmi(70, 1.75):.1f}")

# Invalid call — caught gracefully
try:
    bmi = calculate_bmi(-5, 1.75)
except ValueError as e:
    print(f"Input error: {e}")

print("---")

# Example 2: Re-raising an exception (log it, then pass it up)
def load_data(filename):
    try:
        # Simulate file loading
        if filename.endswith('.xlsx'):
            raise NotImplementedError("Excel files not supported yet")
        print(f"Loaded {filename}")
    except NotImplementedError:
        print(f"Logging: Failed to load {filename}")
        raise    # ← Re-raises the SAME exception to the caller

try:
    load_data("sales.xlsx")
except NotImplementedError as e:
    print(f"Caller caught: {e}")

print("---")

# Example 3: Custom Exception Class
class DataValidationError(Exception):
    """Custom error for data quality issues."""
    def __init__(self, column, message):
        self.column = column
        super().__init__(f"Column '{column}': {message}")

def validate_age(age_value, column_name="age"):
    if not isinstance(age_value, (int, float)):
        raise DataValidationError(column_name, f"Expected number, got {type(age_value).__name__}")
    if age_value < 0 or age_value > 150:
        raise DataValidationError(column_name, f"Value {age_value} out of valid range (0-150)")
    return True

# Test custom exception
test_values = [25, -5, "thirty", 200]
for val in test_values:
    try:
        validate_age(val)
        print(f"  ✅ {val} is valid")
    except DataValidationError as e:
        print(f"  ❌ {e}")
````
