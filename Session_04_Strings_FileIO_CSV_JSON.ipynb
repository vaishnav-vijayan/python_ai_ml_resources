## 1. Strings Deep-Dive

Strings are **immutable sequences of Unicode characters**. Every string operation returns a **new** string — the original is never modified.

### Key Properties Recap
| Property | Description |
|----------|-------------|
| **Immutable** | Cannot change individual characters |
| **Ordered** | Characters maintain their position |
| **Indexed** | Access by position `s[0]`, `s[-1]` |
| **Sliceable** | `s[start:stop:step]` works like lists |
| **Iterable** | `for char in string:` works |
| **Unicode** | Supports all languages, emojis, symbols |

### String Creation Methods
```python
single   = 'hello'           # Single quotes
double   = "hello"           # Double quotes (identical to single)
triple   = '''multi          # Triple quotes for multi-line
line string'''               
raw      = r"C:\new\folder"  # Raw string (backslashes literal)
fstring  = f"Value: {42}"   # f-string (formatted)
escaped  = "tab\there"      # Escape sequences
```

> ⚠️ **Special Case — Single vs Double quotes**: No difference in Python. Use single by convention, switch to double when the string contains an apostrophe: `"it's easy"`. Triple quotes are for multi-line strings and docstrings.

> ⚠️ **Special Case — Raw strings (`r"..."`)**: Backslashes are treated literally. Essential for **Windows file paths** and **regex patterns**: `r"C:\Users\data"` instead of `"C:\\Users\\data"`.

````python
# --- String Basics ---

# Immutability
name = "Python"
print(f"Original: {name}")

try:
    name[0] = 'J'   # Cannot modify in place!
except TypeError as e:
    print(f"TypeError: {e}")

# Create a NEW string instead
new_name = 'J' + name[1:]
print(f"New string: {new_name}")
print(f"Original unchanged: {name}")

# Indexing & slicing (same as lists)
text = "Data Science"
print(f"\ntext[0]:     '{text[0]}'")
print(f"text[-1]:    '{text[-1]}'")
print(f"text[0:4]:   '{text[0:4]}'")
print(f"text[5:]:    '{text[5:]}'")
print(f"text[::-1]:  '{text[::-1]}'")

# String repetition & concatenation
print(f"\n'ha' * 3:    '{'ha' * 3}'")
print(f"'Data' + ' ' + 'Science': '{'Data' + ' ' + 'Science'}'")

# Multi-line strings
query = """
SELECT name, age
FROM students
WHERE grade = 'A'
"""
print(f"SQL query:{query}")
````

### Essential String Methods

**Case Methods:**

| Method | Description | Example |
|--------|-------------|--------|
| `s.upper()` | ALL CAPS | `'hello'` → `'HELLO'` |
| `s.lower()` | all lowercase | `'HELLO'` → `'hello'` |
| `s.title()` | Title Case | `'hello world'` → `'Hello World'` |
| `s.capitalize()` | First letter only | `'hello world'` → `'Hello world'` |
| `s.swapcase()` | Swap case | `'Hello'` → `'hELLO'` |
| `s.casefold()` | Aggressive lowercase | `'Straße'` → `'strasse'` |

**Search & Check Methods:**

| Method | Returns | Raises? |
|--------|---------|--------|
| `s.find(sub)` | Index or `-1` | Never |
| `s.index(sub)` | Index | `ValueError` if missing |
| `s.rfind(sub)` | Last index or `-1` | Never |
| `s.count(sub)` | Number of occurrences | Never |
| `s.startswith(prefix)` | `True`/`False` | Never |
| `s.endswith(suffix)` | `True`/`False` | Never |
| `'sub' in s` | `True`/`False` | Never |

**Validation Methods (return `True`/`False`):**

| Method | Checks for |
|--------|------------|
| `s.isalpha()` | Only letters |
| `s.isdigit()` | Only digits |
| `s.isalnum()` | Letters or digits |
| `s.isspace()` | Only whitespace |
| `s.isupper()` | All uppercase |
| `s.islower()` | All lowercase |
| `s.isnumeric()` | Numeric (incl. ½, ², etc.) |

> ⚠️ **Special Case — `find()` vs `index()`**: Use `find()` when you're not sure the substring exists (returns `-1`). Use `index()` when absence is an error you want to catch.

> ⚠️ **Special Case — `casefold()` vs `lower()`**: For case-insensitive comparisons, `casefold()` is more aggressive and handles non-English characters correctly. `'ß'.casefold()` → `'ss'`, while `'ß'.lower()` → `'ß'`.

> **Data Science relevance**: Case normalization (`lower()`/`casefold()`) is the first step in text preprocessing. `startswith()`/`endswith()` filter file types. `isdigit()` validates data.

````python
# --- String Methods ---

text = "  Hello, World! Welcome to Python Data Science.  "
print(f"Original: '{text}'")

# Case methods
print(f"\nupper():      '{text.upper()}'")
print(f"lower():      '{text.lower()}'")
print(f"title():      '{text.title()}'")
print(f"capitalize(): '{text.capitalize()}'")

# Search methods
print(f"\nfind('World'):    {text.find('World')}")
print(f"find('Missing'):  {text.find('Missing')}")      # -1 (not found)
print(f"count('o'):       {text.count('o')}")
print(f"startswith(' '):  {text.startswith(' ')}")
print(f"'Python' in text: {'Python' in text}")

# Validation methods
print(f"\n'hello'.isalpha(): {'hello'.isalpha()}")
print(f"'12345'.isdigit(): {'12345'.isdigit()}")
print(f"'abc123'.isalnum(): {'abc123'.isalnum()}")
print(f"'  '.isspace(): {'  '.isspace()}")

# Case-insensitive comparison
user_input = "PYTHON"
print(f"\nCase-insensitive compare:")
print(f"  '{user_input}'.lower() == 'python': {user_input.lower() == 'python'}")

# casefold for non-English
german = "Straße"
print(f"\n'{german}'.lower():    '{german.lower()}'")
print(f"'{german}'.casefold(): '{german.casefold()}'")
````

### Modifying Strings — `strip()`, `replace()`, `split()`, `join()`

These are the **four most important string methods for data cleaning**.

| Method | Description | Data Science Use |
|--------|-------------|------------------|
| `s.strip()` | Remove leading/trailing whitespace | Clean CSV column values |
| `s.lstrip()` / `s.rstrip()` | Remove from left/right only | Remove specific padding |
| `s.strip(chars)` | Remove specific characters | Remove `$`, `%`, quotes |
| `s.replace(old, new)` | Replace ALL occurrences | Fix inconsistent data |
| `s.replace(old, new, count)` | Replace first `count` only | Targeted replacement |
| `s.split()` | Split by whitespace → list | Tokenize text |
| `s.split(sep)` | Split by separator | Parse CSV, logs |
| `s.split(sep, maxsplit)` | Split limited times | Parse structured strings |
| `s.rsplit(sep, maxsplit)` | Split from right | Get file extension |
| `sep.join(list)` | Join list items with sep | Rebuild strings |

> ⚠️ **Special Case — `split()` with no argument vs `split(' ')`:**
> - `'a  b  c'.split()` → `['a', 'b', 'c']` (splits on ANY whitespace, removes empties)
> - `'a  b  c'.split(' ')` → `['a', '', 'b', '', 'c']` (splits on single space, keeps empties)
> 
> Always use `split()` (no argument) for tokenizing text.

> ⚠️ **Special Case — `strip()` removes CHARACTERS, not substrings!**
> ```python
> 'abcba'.strip('ab')  # → 'c' (removes any 'a' or 'b' from both ends)
> 'hello world'.strip('held')  # → 'o wor' (removes h,e,l,d from ends)
> ```

````python
# --- strip(), replace(), split(), join() ---

# strip — cleaning whitespace
messy = "   Alice   "
print(f"Original: '{messy}'")
print(f"strip():  '{messy.strip()}'")
print(f"lstrip(): '{messy.lstrip()}'")
print(f"rstrip(): '{messy.rstrip()}'")

# strip specific characters (data cleaning)
price = "$1,299.99"
clean_price = price.strip('$').replace(',', '')
print(f"\nPrice: '{price}' → cleaned: {float(clean_price)}")

# replace — fix inconsistent data
data = "Male, female, MALE, Female, male"
normalized = data.lower().replace('male', 'M').replace('feM', 'F')
# Better approach:
values = [v.strip().lower() for v in data.split(',')]
normalized2 = ['M' if v == 'male' else 'F' for v in values]
print(f"\nRaw: {data}")
print(f"Normalized: {normalized2}")

# split — tokenizing text
sentence = "Python is great for data science"
words = sentence.split()
print(f"\nWords: {words}")
print(f"Word count: {len(words)}")

# split() vs split(' ')
spaced = "a  b  c"
print(f"\nsplit():    {spaced.split()}")       # ['a', 'b', 'c']
print(f"split(' '): {spaced.split(' ')}")      # ['a', '', 'b', '', 'c']

# split with separator
csv_line = "Alice,30,Engineer,NYC"
fields = csv_line.split(',')
print(f"\nCSV fields: {fields}")

# split with maxsplit
log = "2024-01-15 ERROR Database connection failed: timeout"
parts = log.split(' ', 2)   # Split into 3 parts max
print(f"Log parts: {parts}")

# join — the opposite of split
words = ['Python', 'Data', 'Science']
print(f"\nJoin with space: '{' '.join(words)}'")
print(f"Join with comma: '{', '.join(words)}'")
print(f"Join with arrow: '{' → '.join(words)}'")

# Common pattern: clean then rejoin
messy_csv = " Alice , 30 , NYC "
clean_csv = ','.join(field.strip() for field in messy_csv.split(','))
print(f"\nCleaned CSV: '{clean_csv}'")
````

### String Formatting — f-strings, `format()`, `%`

Python has 3 ways to format strings. **f-strings are the modern standard** (Python 3.6+).

| Method | Syntax | Recommendation |
|--------|--------|----------------|
| **f-string** | `f"Hello {name}"` | ✅ Always use this |
| `.format()` | `"Hello {}".format(name)` | Legacy code |
| `%` formatting | `"Hello %s" % name` | Very old (C-style) |

### f-string Format Specifiers

```python
f"{value:format_spec}"
```

| Specifier | Meaning | Example | Result |
|-----------|---------|---------|--------|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `'3.14'` |
| `:,` | Thousands separator | `f"{1000000:,}"` | `'1,000,000'` |
| `:.2%` | Percentage | `f"{0.856:.2%}"` | `'85.60%'` |
| `:>10` | Right-align, width 10 | `f"{'hi':>10}"` | `'        hi'` |
| `:<10` | Left-align, width 10 | `f"{'hi':<10}"` | `'hi        '` |
| `:^10` | Center, width 10 | `f"{'hi':^10}"` | `'    hi    '` |
| `:08d` | Zero-pad to 8 digits | `f"{42:08d}"` | `'00000042'` |
| `:+.1f` | Show sign always | `f"{3.5:+.1f}"` | `'+3.5'` |
| `:#x` | Hex with prefix | `f"{255:#x}"` | `'0xff'` |
| `:e` | Scientific notation | `f"{0.001:e}"` | `'1.000000e-03'` |

> ⚠️ **Special Case — f-strings can contain ANY expression**: `f"{2**10}"`, `f"{name.upper()}"`, `f"{'pass' if score >= 50 else 'fail'}"`. But keep them simple — complex logic belongs outside the f-string.

> **Data Science relevance**: Format specifiers are essential for clean reports — displaying percentages, large numbers with commas, aligned tables.

````python
# --- String Formatting ---

name = "Alice"
score = 0.9234
total = 1234567

# f-string (recommended)
print(f"Student: {name}, Score: {score:.2%}, Total: {total:,}")

# Useful format specifiers for data science
pi = 3.14159265
print(f"\n2 decimals:    {pi:.2f}")
print(f"Scientific:    {0.000123:e}")
print(f"Percentage:    {0.856:.1%}")
print(f"Thousands:     {9876543:,}")
print(f"Zero-padded:   {42:08d}")
print(f"Right-aligned: '{name:>15}'")
print(f"Left-aligned:  '{name:<15}'")
print(f"Centered:      '{name:^15}'")

# Expressions inside f-strings
values = [10, 20, 30, 40]
print(f"\nSum: {sum(values)}, Avg: {sum(values)/len(values):.1f}")

# Alignment for tables
print(f"\n{'Name':<10} {'Score':>8} {'Grade':^8}")
print(f"{'-'*10} {'-'*8} {'-'*8}")
students = [('Alice', 92, 'A'), ('Bob', 78, 'B+'), ('Charlie', 95, 'A+')]
for name, score, grade in students:
    print(f"{name:<10} {score:>8} {grade:^8}")

# Debugging with f-string (Python 3.8+)
x = 42
print(f"\n{x = }")            # Prints: x = 42
print(f"{len(students) = }")   # Prints: len(students) = 3
````

### Regular Expressions (Regex) Basics

Regular expressions are **patterns** for matching text. Python's `re` module provides regex support.

### Most Common Patterns

| Pattern | Matches | Example |
|---------|---------|--------|
| `\d` | Any digit (0-9) | `'abc123'` → `['1','2','3']` |
| `\D` | Non-digit | `'abc123'` → `['a','b','c']` |
| `\w` | Word char (letter/digit/_) | `'hi-5!'` → `['h','i','5']` |
| `\s` | Whitespace | `'a b\tc'` → `[' ','\t']` |
| `.` | Any character (except newline) | `'a1!'` → `['a','1','!']` |
| `+` | One or more | `\d+` matches `'123'` as one match |
| `*` | Zero or more | `\d*` matches `''` and `'123'` |
| `?` | Zero or one | `colou?r` matches `'color'` and `'colour'` |
| `{n}` | Exactly n times | `\d{4}` matches `'2024'` |
| `[abc]` | Any of a, b, c | `[aeiou]` matches vowels |
| `^` | Start of string | `^Hello` matches `'Hello world'` |
| `$` | End of string | `world$` matches `'Hello world'` |
| `\|` | OR | `cat\|dog` matches either |
| `()` | Capture group | `(\d+)-(\d+)` captures parts |

### Key `re` Functions

| Function | Returns | Description |
|----------|---------|-------------|
| `re.search(pattern, text)` | Match object or `None` | Find first match |
| `re.findall(pattern, text)` | List of strings | Find ALL matches |
| `re.sub(pattern, repl, text)` | New string | Replace all matches |
| `re.split(pattern, text)` | List of strings | Split by pattern |
| `re.match(pattern, text)` | Match object or `None` | Match at START only |

> ⚠️ **Special Case — Always use raw strings for regex**: `re.search(r'\d+', text)`. Without `r`, Python interprets backslashes as escape sequences before regex sees them.

> **Data Science relevance**: Extracting numbers from text, validating emails/phones, cleaning column names, parsing log files, text preprocessing for NLP.

````python
# --- Regular Expressions ---
import re

text = "Order #1234 placed on 2024-01-15 for $99.99. Contact: alice@email.com"

# findall — extract all matches
numbers = re.findall(r'\d+', text)
print(f"All numbers: {numbers}")

# Extract specific patterns
date = re.search(r'\d{4}-\d{2}-\d{2}', text)
print(f"Date found: {date.group() if date else 'Not found'}")

price = re.search(r'\$([\d.]+)', text)
print(f"Price: {price.group(1) if price else 'Not found'}")

email = re.search(r'[\w.]+@[\w.]+', text)
print(f"Email: {email.group() if email else 'Not found'}")

# sub — replace with pattern
# Mask email addresses
masked = re.sub(r'[\w.]+@[\w.]+', '[REDACTED]', text)
print(f"\nMasked: {masked}")

# Clean column names (common in data science)
columns = ['First Name', 'Last  Name', 'Email-Address', 'Phone #']
clean_cols = [re.sub(r'[^\w]', '_', col.strip().lower()) for col in columns]
# Remove consecutive underscores
clean_cols = [re.sub(r'_+', '_', col).strip('_') for col in clean_cols]
print(f"\nOriginal columns: {columns}")
print(f"Clean columns:    {clean_cols}")

# Validate patterns
def is_valid_email(email):
    pattern = r'^[\w.+-]+@[\w-]+\.[\w.-]+$'
    return bool(re.match(pattern, email))

emails = ['alice@email.com', 'invalid@', 'bob@company.co.uk', '@missing.com']
print("\nEmail validation:")
for e in emails:
    print(f"  {e:25s} → {'✅ Valid' if is_valid_email(e) else '❌ Invalid'}")

# Extract multiple groups
log = "2024-01-15 14:30:22 ERROR Database timeout after 30s"
match = re.match(r'(\S+) (\S+) (\w+) (.+)', log)
if match:
    date, time, level, message = match.groups()
    print(f"\nLog parsed: date={date}, time={time}, level={level}")
    print(f"  message: {message}")
````

## 2. File I/O — Reading & Writing Files

Python's built-in `open()` function handles all file operations.

### The `open()` Function
```python
file = open(filename, mode, encoding='utf-8')
```

### File Modes

| Mode | Description | Creates file? | Overwrites? |
|------|-------------|:---:|:---:|
| `'r'` | Read (default) | ❌ | ❌ |
| `'w'` | Write | ✅ | ✅ **Destroys existing!** |
| `'a'` | Append | ✅ | ❌ (adds to end) |
| `'x'` | Exclusive create | ✅ | ❌ (error if exists) |
| `'r+'` | Read & write | ❌ | ❌ |
| `'rb'` | Read binary | ❌ | ❌ |
| `'wb'` | Write binary | ✅ | ✅ |

### Always Use `with` Statement (Context Manager)

```python
# ✅ GOOD — file automatically closed, even if error occurs
with open('file.txt', 'r', encoding='utf-8') as f:
    content = f.read()
# File is closed here

# ❌ BAD — must remember to close, may leak if error occurs
f = open('file.txt', 'r')
content = f.read()
f.close()           # Easy to forget!
```

> ⚠️ **Special Case — `'w'` mode DESTROYS existing files!** It deletes all content before writing. Use `'a'` (append) to add to an existing file.

> ⚠️ **Special Case — Always specify `encoding='utf-8'`**: Without it, Python uses the system default (varies by OS). This causes bugs when moving code between Windows and Mac/Linux.

> ⚠️ **Special Case — `with` handles cleanup even during exceptions**: If an error occurs inside the `with` block, the file is still properly closed. This is crucial for database connections, network sockets, etc.

````python
# --- Writing Files ---

# Write a text file
with open('sample.txt', 'w', encoding='utf-8') as f:
    f.write('Line 1: Hello, File I/O!\n')
    f.write('Line 2: Python is great.\n')
    f.write('Line 3: Data Science rocks.\n')

print("✅ sample.txt created")

# Write multiple lines at once
lines = [
    'Alice,92,A\n',
    'Bob,78,B+\n',
    'Charlie,95,A+\n',
    'Diana,85,B+\n'
]

with open('students.txt', 'w', encoding='utf-8') as f:
    f.write('Name,Score,Grade\n')   # Header
    f.writelines(lines)             # Write all lines

print("✅ students.txt created")

# Append to existing file
with open('sample.txt', 'a', encoding='utf-8') as f:
    f.write('Line 4: Appended line!\n')

print("✅ Appended to sample.txt")
````

````python
# --- Reading Files ---

# Method 1: read() — entire file as one string
with open('sample.txt', 'r', encoding='utf-8') as f:
    content = f.read()
print("--- read() ---")
print(content)

# Method 2: readline() — one line at a time
with open('sample.txt', 'r', encoding='utf-8') as f:
    first_line = f.readline()      # Reads one line
    second_line = f.readline()     # Reads next line
print("--- readline() ---")
print(f"First:  {first_line.strip()}")
print(f"Second: {second_line.strip()}")

# Method 3: readlines() — all lines as a list
with open('sample.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()
print("--- readlines() ---")
print(f"Lines: {lines}")
print(f"Count: {len(lines)} lines")

# Method 4: Iterate line-by-line (BEST for large files — memory efficient!)
print("\n--- for loop (best practice) ---")
with open('students.txt', 'r', encoding='utf-8') as f:
    for line_num, line in enumerate(f, 1):
        print(f"  Line {line_num}: {line.strip()}")
````

### Reading Methods Comparison

| Method | Returns | Memory | Use When |
|--------|---------|--------|----------|
| `f.read()` | Entire file as string | Loads ALL into memory | Small files |
| `f.read(n)` | First `n` characters | Controlled | Peeking at content |
| `f.readline()` | One line | Minimal | Processing line by line |
| `f.readlines()` | List of all lines | Loads ALL into memory | Need random access to lines |
| `for line in f:` | One line per iteration | **Minimal (generator)** | **Large files (recommended)** |

> ⚠️ **Special Case — `readlines()` keeps `\n`**: Each line includes the trailing newline. Use `line.strip()` or `line.rstrip('\n')` to remove it.

> ⚠️ **Special Case — File pointer position**: After `read()`, the file pointer is at the end. Calling `read()` again returns `''`. Use `f.seek(0)` to reset to the beginning.

> **Data Science relevance**: For multi-GB log files or data dumps, always use `for line in f:` — it reads one line at a time without loading the entire file into memory.

````python
# --- Practical File I/O Patterns ---

# Pattern 1: Read, process, write
with open('students.txt', 'r', encoding='utf-8') as f:
    header = f.readline().strip()
    records = []
    for line in f:
        parts = line.strip().split(',')
        name, score, grade = parts[0], int(parts[1]), parts[2]
        records.append({'name': name, 'score': score, 'grade': grade})

print(f"Header: {header}")
print(f"Records: {records}")
avg_score = sum(r['score'] for r in records) / len(records)
print(f"Average score: {avg_score:.1f}")

# Pattern 2: Check if file exists before reading
import os

filename = 'nonexistent.txt'
if os.path.exists(filename):
    with open(filename, 'r') as f:
        print(f.read())
else:
    print(f"\n'{filename}' does not exist")

# Pattern 3: File pointer demo
with open('sample.txt', 'r', encoding='utf-8') as f:
    first_read = f.read()
    second_read = f.read()        # Empty! Pointer at end.
    print(f"\nSecond read is empty: '{second_read}'")
    
    f.seek(0)                      # Reset to beginning
    third_read = f.read(10)        # Read first 10 chars
    print(f"After seek(0): '{third_read}'")

# Pattern 4: Count lines, words, characters
with open('sample.txt', 'r', encoding='utf-8') as f:
    content = f.read()
    
lines = content.strip().split('\n')
words = content.split()
chars = len(content)
print(f"\nFile stats: {len(lines)} lines, {len(words)} words, {chars} chars")
````

## 3. Working with CSV Files

**CSV** (Comma-Separated Values) is the most common data format in data science. Python provides two ways to handle CSV:

| Method | When to Use |
|--------|------------|
| `csv` module (built-in) | Simple CSV, no external dependencies |
| `pandas.read_csv()` | Data analysis (we'll learn this in Session 9) |

### CSV Challenges
- Values may contain commas: `"Smith, John"` (needs quoting)
- Different delimiters: tab (`\t`), semicolon (`;`), pipe (`|`)
- Header row may or may not exist
- Encoding issues (UTF-8 vs Latin-1 vs Windows-1252)
- Inconsistent quoting

> ⚠️ **Special Case — Never parse CSV manually with `split(',')`!** It breaks when values contain commas. Always use the `csv` module or Pandas.

````python
# --- Writing CSV with csv module ---
import csv

# Write CSV with csv.writer
header = ['Name', 'Age', 'City', 'Salary']
data = [
    ['Alice', 30, 'New York', 85000],
    ['Bob', 25, 'San Francisco', 92000],
    ['Charlie', 35, 'Chicago', 78000],
    ['Diana', 28, 'Boston', 95000],
    ['Eve', 32, 'Seattle', 88000],
]

with open('employees.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(header)      # Write header
    writer.writerows(data)       # Write all rows

print("✅ employees.csv created")

# Verify by reading raw
with open('employees.csv', 'r', encoding='utf-8') as f:
    print(f.read())
````

````python
# --- Reading CSV with csv.reader ---
import csv

# Method 1: csv.reader — returns lists
print("--- csv.reader ---")
with open('employees.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader)        # Skip and store header
    print(f"Header: {header}")
    
    for row in reader:
        name, age, city, salary = row
        print(f"  {name}, age {age}, lives in {city}, earns ${int(salary):,}")
````

````python
# --- Reading CSV with csv.DictReader --- (MORE USEFUL!)
import csv

# Method 2: csv.DictReader — returns dictionaries (access by column name)
print("--- csv.DictReader ---")
with open('employees.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    
    employees = list(reader)   
    print(employees)  # Convert to list of dicts

for emp in employees:
    print(f"  {emp['Name']}: ${int(emp['Salary']):,} in {emp['City']}")

# Easy to compute statistics
salaries = [int(emp['Salary']) for emp in employees]
print(f"\nSalary stats:")
print(f"  Average: ${sum(salaries)/len(salaries):,.0f}")
print(f"  Min:     ${min(salaries):,}")
print(f"  Max:     ${max(salaries):,}")

# Filter employees
high_earners = [e['Name'] for e in employees if int(e['Salary']) > 85000]
print(f"  Earning > $85K: {high_earners}")
````

````python
# --- Writing CSV with DictWriter ---
import csv

# DictWriter — write from dictionaries
fieldnames = ['Name', 'Department', 'Score']
records = [
    {'Name': 'Alice', 'Department': 'Engineering', 'Score': 92},
    {'Name': 'Bob', 'Department': 'Marketing', 'Score': 85},
    {'Name': 'Charlie', 'Department': 'Engineering', 'Score': 95},
]

with open('departments.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()         # Write column names
    writer.writerows(records)

print("✅ departments.csv created")

# Handling commas in values (csv module handles quoting automatically!)
tricky_data = [
    ['Smith, John', 'New York, NY', '85000'],
    ['O\'Brien, Alice', 'San Francisco, CA', '92000'],
]

with open('tricky.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['Name', 'Location', 'Salary'])
    writer.writerows(tricky_data)

# Read it back — commas in values are handled correctly
print("\n--- Tricky CSV with commas in values ---")
with open('tricky.csv', 'r', encoding='utf-8') as f:
    print(f"Raw file:\n{f.read()}")

with open('tricky.csv', 'r', encoding='utf-8') as f:
    for row in csv.reader(f):
        print(f"Parsed: {row}")
````

### `csv.reader` vs `csv.DictReader` Comparison

| Feature | `csv.reader` | `csv.DictReader` |
|---------|-------------|------------------|
| Returns | Lists `['Alice', '30']` | Dicts `{'Name': 'Alice', 'Age': '30'}` |
| Access | By index `row[0]` | By name `row['Name']` |
| Readable | ❌ Less | ✅ More |
| Header handling | Manual `next(reader)` | Automatic (first row) |
| Speed | Slightly faster | Slightly slower |
| Use when | Column order is fixed | Column names might change |

> ⚠️ **Special Case — `newline=''` on Windows**: Always pass `newline=''` when opening CSV files for writing on Windows. Without it, you get double-spaced output (extra blank lines).

> ⚠️ **Special Case — All CSV values are strings!** `csv.reader` returns everything as strings. You must convert: `int(row[1])`, `float(row[3])`. DictReader has the same behavior.

## 4. Working with JSON Data

**JSON** (JavaScript Object Notation) is the standard format for **APIs**, **web data**, and **configuration files**.

### JSON ↔ Python Type Mapping

| JSON Type | Python Type |
|-----------|------------|
| `object` `{"key": "value"}` | `dict` |
| `array` `[1, 2, 3]` | `list` |
| `string` `"hello"` | `str` |
| `number` (int) `42` | `int` |
| `number` (float) `3.14` | `float` |
| `true` / `false` | `True` / `False` |
| `null` | `None` |

### Key Functions

| Function | Direction | Description |
|----------|-----------|-------------|
| `json.dumps(obj)` | Python → JSON string | Serialize to string |
| `json.dump(obj, file)` | Python → JSON file | Write to file |
| `json.loads(string)` | JSON string → Python | Parse from string |
| `json.load(file)` | JSON file → Python | Read from file |

> **Memory trick**: `dumps` = dump **s**tring, `loads` = load **s**tring. Without `s` = file operations.

> ⚠️ **Special Case — JSON keys MUST be strings**: `{1: 'one'}` in Python becomes `{"1": "one"}` in JSON (int key converted to string).

> ⚠️ **Special Case — JSON has no comments**: Unlike Python, JSON doesn't support `#` or `//` comments. Use JSONC or YAML if you need comments.

> **Data Science relevance**: REST APIs return JSON. Config files for ML pipelines. NoSQL databases (MongoDB) store JSON documents.

````python
# --- JSON Basics ---
import json

# Python dict → JSON string (serialization)
student = {
    'name': 'Alice',
    'age': 20,
    'courses': ['Python', 'Statistics', 'ML'],
    'grades': {'Python': 92, 'Statistics': 88, 'ML': 95},
    'active': True,
    'advisor': None
}

# Convert to JSON string
json_str = json.dumps(student)
print(f"JSON string:\n{json_str}")

# Pretty-printed JSON (much more readable!)
pretty = json.dumps(student, indent=2)
print(f"\nPretty JSON:\n{pretty}")

# JSON string → Python dict (deserialization)
parsed = json.loads(json_str)
print(f"\nParsed Python object:\n{parsed}")
print(f"\nParsed type: {type(parsed)}")
print(f"Name: {parsed['name']}")
print(f"Courses: {parsed['courses']}")
print(f"ML grade: {parsed['grades']['ML']}")
print(f"Advisor: {parsed['advisor']}")   # None became null in JSON
````

````python
# --- JSON Files: Read & Write ---
import json

# Write JSON to file
data = {
    'company': 'TechCorp',
    'employees': [
        {'name': 'Alice', 'role': 'Data Scientist', 'salary': 95000},
        {'name': 'Bob', 'role': 'ML Engineer', 'salary': 105000},
        {'name': 'Charlie', 'role': 'Analyst', 'salary': 72000}
    ],
    'departments': ['Engineering', 'Analytics', 'Marketing'],
    'founded': 2015
}

# Write to file (json.dump — no 's'!)
with open('company.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, indent=2)

print("✅ company.json created")

# Read from file (json.load — no 's'!)
with open('company.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)
print(f"\nLoaded JSON data:\n{loaded}")
print(f"\nLoaded type: {type(loaded)}")
print(f"\nCompany: {loaded['company']}")
print(f"Employees:")
for emp in loaded['employees']:
    print(f"  {emp['name']}: {emp['role']} (${emp['salary']:,})")

# Calculate stats from JSON data
salaries = [e['salary'] for e in loaded['employees']]
print(f"\nAverage salary: ${sum(salaries)/len(salaries):,.0f}")
````

````python
# --- Handling JSON Edge Cases ---
import json

# JSON keys must be strings
python_dict = {1: 'one', 2: 'two', (3, 4): 'tuple_key'}

# Tuples can't be JSON keys — handle the error
try:
    json_str = json.dumps(python_dict)
except TypeError as e:
    print(f"TypeError: {e}")

# Int keys get converted to strings
simple = {1: 'one', 2: 'two'}
json_str = json.dumps(simple)
parsed = json.loads(json_str)
print(f"\nOriginal keys: {list(simple.keys())}")
print(f"After JSON roundtrip: {list(parsed.keys())}")
print(f"Key types changed: int → {type(list(parsed.keys())[0]).__name__}!")

# Handling dates (not supported natively)
from datetime import datetime

event = {'name': 'Conference', 'date': datetime.now()}
try:
    json.dumps(event)
except TypeError as e:
    print(f"\nTypeError: {e}")

# Solution: convert dates to strings
event['date'] = event['date'].isoformat()
print(f"With ISO date: {json.dumps(event, indent=2)}")

# Or use a custom serializer
def json_serializer(obj):
    if isinstance(obj, datetime):
        return obj.isoformat()
    raise TypeError(f"Object of type {type(obj)} is not JSON serializable")

event2 = {'name': 'Workshop', 'date': datetime.now()}
print(f"\nCustom serializer: {json.dumps(event2, default=json_serializer, indent=2)}")
````

````python
# --- Nested JSON: Parsing API-Style Data ---
import json

# Simulate API response (common in data science)
api_response = '''
{
    "status": "success",
    "count": 3,
    "results": [
        {
            "id": 1,
            "name": "Widget A",
            "price": 29.99,
            "tags": ["sale", "popular"],
            "specs": {"weight": 0.5, "color": "blue"}
        },
        {
            "id": 2,
            "name": "Widget B",
            "price": 49.99,
            "tags": ["new"],
            "specs": {"weight": 0.8, "color": "red"}
        },
        {
            "id": 3,
            "name": "Widget C",
            "price": 19.99,
            "tags": ["sale"],
            "specs": {"weight": 0.3, "color": "green"}
        }
    ]
}
'''

# Parse the JSON
data = json.loads(api_response)
print(f"Parsed API response:\n{data}")
print(f"\nParsed type: {type(data)}")
print(f"Status: {data['status']}")
print(f"Results count: {data['count']}")

# Navigate nested data
print(f"\nAll products:")
for product in data['results']:
    print(f"  {product['name']}: ${product['price']} ({product['specs']['color']})")

# Filter: products on sale
on_sale = [p for p in data['results'] if 'sale' in p['tags']]
print(f"\nOn sale:")
for p in on_sale:
    print(f"  {p['name']}: ${p['price']}")

# Safe nested access (common pattern)
def safe_get(data, *keys, default=None):
    """Safely navigate nested dicts/lists."""
    for key in keys:
        try:
            data = data[key]
        except (KeyError, IndexError, TypeError):
            return default
    return data

# Safe access demo
print(f"\nSafe access:")
print(f"  Exists:  {safe_get(data, 'results', 0, 'name')}")
print(f"  Missing: {safe_get(data, 'results', 99, 'name', default='N/A')}")
````

### CSV vs JSON Comparison

| Feature | CSV | JSON |
|---------|-----|------|
| Structure | Flat (rows/columns) | Nested (hierarchical) |
| Data types | Everything is text | Strings, numbers, booleans, null |
| Readability | Good for tabular data | Good for nested/tree data |
| File size | Smaller | Larger (due to keys) |
| Parsing speed | Faster | Slower |
| Use case | Datasets, spreadsheets | APIs, configs, nested data |
| Python module | `csv` → Pandas | `json` → `requests` |
| Comments | No (by default) | No |

> **When to use which:**
> - **CSV**: Tabular data (rows/columns), export from databases, input for Pandas
> - **JSON**: API responses, config files, nested/hierarchical data, NoSQL databases

## 5. Bonus: `pathlib` — Modern File Path Handling

The `pathlib` module (Python 3.4+) provides an **object-oriented** way to handle file paths. It's cleaner than `os.path`.

| `os.path` (old) | `pathlib` (modern) |
|-----------------|-------------------|
| `os.path.join('a', 'b')` | `Path('a') / 'b'` |
| `os.path.exists(path)` | `path.exists()` |
| `os.path.isfile(path)` | `path.is_file()` |
| `os.path.basename(path)` | `path.name` |
| `os.path.splitext(path)` | `path.suffix` |
| `os.makedirs(path)` | `path.mkdir(parents=True)` |
| `os.listdir(path)` | `path.iterdir()` |
| `glob.glob('*.csv')` | `path.glob('*.csv')` |

````python
# --- pathlib Demo ---
from pathlib import Path

# Current working directory
cwd = Path.cwd()
print(f"Current directory: {cwd}")

# Build paths with / operator (works on Windows too!)
data_dir = cwd / 'data'
file_path = data_dir / 'results.csv'
print(f"Data dir: {data_dir}")
print(f"File path: {file_path}")

# Path properties
p = Path('employees.csv')
print(f"\nPath properties for '{p}':")
print(f"  name:   {p.name}")       # 'employees.csv'
print(f"  stem:   {p.stem}")       # 'employees'
print(f"  suffix: {p.suffix}")     # '.csv'
print(f"  exists: {p.exists()}")
print(f"  is_file: {p.is_file()}")

# Read/write with pathlib (no open() needed!)
p = Path('pathlib_demo.txt')
p.write_text('Hello from pathlib!\nLine 2.', encoding='utf-8')
content = p.read_text(encoding='utf-8')
print(f"\nRead with pathlib: {content}")

# Find all CSV files in current directory
csv_files = list(Path('.').glob('*.csv'))
print(f"\nCSV files found: {[f.name for f in csv_files]}")

# List all files
print(f"\nAll files in directory:")
for item in sorted(Path('.').iterdir()):
    icon = '📁' if item.is_dir() else '📄'
    print(f"  {icon} {item.name}")

# Cleanup demo file
p.unlink()  # Delete the file
print(f"\n✅ Cleaned up {p.name}")
````
