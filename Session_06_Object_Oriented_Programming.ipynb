## 1. Classes and Objects

A **class** is a blueprint for creating objects. An **object** is an instance of a class.

### Why Use Classes?

| Without Classes | With Classes |
|-----------------|-------------|
| `person_name = 'Alice'` `person_age = 30` `person_salary = 85000` | `person = Person('Alice', 30, 85000)` |
| Related data scattered everywhere | Data grouped logically |
| Hard to perform actions on data | Methods bundle behavior with data |
| Reuse patterns by copying code | Reuse through inheritance |
| Unclear what data types are expected | Type is explicit and enforced |

### Basic Class Structure

```python
class Person:
    """Represents a person with name and age."""
    
    def __init__(self, name, age):
        """Initialize a person."""
        self.name = name          # Attribute
        self.age = age
    
    def greet(self):
        """Method — an action the object can perform."""
        return f"Hello, I'm {self.name}"
    
    def have_birthday(self):
        """Increment age."""
        self.age += 1

# Create an object (instance)
person = Person('Alice', 30)
print(person.greet())        # Hello, I'm Alice
print(person.age)            # 30
person.have_birthday()
print(person.age)            # 31
```

### Key Terminology

| Term | Definition | Example |
|------|-----------|----------|
| **Class** | Blueprint for objects | `class Person:` |
| **Object/Instance** | A specific thing created from class | `alice = Person('Alice', 30)` |
| **Attribute** | Data belonging to object | `person.name`, `person.age` |
| **Method** | Function belonging to object | `person.greet()` |
| **`self`** | Reference to current object | Used in method definitions |
| **`__init__`** | Constructor; runs when object created | Initialize attributes |

> ⚠️ **Special Case — `self` is always first parameter**: Every method gets `self` automatically. You don't pass it when calling: `person.greet()` (not `person.greet(person)`). Python handles this.

> **Data Science relevance**: Scikit-Learn models are classes. `LogisticRegression` is a class; `model = LogisticRegression()` creates an object.

````python
# --- Simple Class Example ---

class Student:
    """Represents a student with grades."""
    
    def __init__(self, name, student_id):
        """Initialize student."""
        self.name = name
        self.student_id = student_id
        self.grades = []
    
    def add_grade(self, grade):
        """Add a grade."""
        if 0 <= grade <= 100:
            self.grades.append(grade)
        else:
            print("Invalid grade (must be 0-100)")
    
    def get_average(self):
        """Calculate average grade."""
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)
    
    def get_letter_grade(self):
        """Return letter grade based on average."""
        avg = self.get_average()
        if avg >= 90:
            return 'A'
        elif avg >= 80:
            return 'B'
        elif avg >= 70:
            return 'C'
        else:
            return 'F'

# Create objects
alice = Student('Alice', 'S001')
bob = Student('Bob', 'S002')

# Add grades
alice.add_grade(92)
alice.add_grade(95)
alice.add_grade(88)

bob.add_grade(78)
bob.add_grade(82)
bob.add_grade(79)

# Access attributes and call methods
print(f"Student: {alice.name}")
print(f"Grades: {alice.grades}")
print(f"Average: {alice.get_average():.1f}")
print(f"Letter Grade: {alice.get_letter_grade()}")

print(f"\nStudent: {bob.name}")
print(f"Average: {bob.get_average():.1f}")
print(f"Letter Grade: {bob.get_letter_grade()}")

# Each object is independent
print(f"\nAlice and Bob have different data:")
print(f"  alice.grades = {alice.grades}")
print(f"  bob.grades = {bob.grades}")
````

## 2. Special Methods (`dunder` Methods)

**Dunder methods** (double underscore) are special methods that Python calls automatically. They customize how your object behaves.

### Most Common Special Methods

| Method | Called When | Example |
|--------|-------------|----------|
| `__init__` | Object is created | `Person('Alice', 30)` |
| `__str__` | `print(obj)` or `str(obj)` | `str(alice)` → `'Alice, 30'` |
| `__repr__` | Interactive mode or `repr(obj)` | `repr(alice)` → `'Person(Alice, 30)'` |
| `__len__` | `len(obj)` | `len(team)` |
| `__getitem__` | `obj[index]` | `data[0]` |
| `__setitem__` | `obj[index] = value` | `data[0] = 100` |
| `__eq__` | `obj1 == obj2` | `alice == bob` |
| `__lt__`, `__gt__`, etc. | Comparisons | `alice < bob` |
| `__add__` | `obj1 + obj2` | `date1 + date2` |
| `__call__` | `obj()` | `model(features)` |

### `__init__` and `__str__`

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        """Human-readable representation for print()."""
        return f"{self.name}, age {self.age}"
    
    def __repr__(self):
        """Developer-friendly representation for debugging."""
        return f"Person('{self.name}', {self.age})"

alice = Person('Alice', 30)
print(alice)        # Uses __str__: "Alice, age 30"
repr(alice)         # Uses __repr__: "Person('Alice', 30)"
```

> ⚠️ **Special Case — `__str__` vs `__repr__`**:
> - `__str__` is for **end users** — should be friendly and readable
> - `__repr__` is for **developers** — should show exactly how to recreate the object
> 
> If only `__repr__` is defined, Python uses it for both `print()` and `repr()`.

````python
# --- Special Methods: __str__ and __repr__ ---

class Product:
    """Represents a product with price."""
    
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def __str__(self):
        """User-friendly display."""
        return f"{self.name}: ${self.price:.2f}"
    
    def __repr__(self):
        """Developer-friendly display."""
        return f"Product('{self.name}', {self.price})"

laptop = Product('MacBook Pro', 1299.99)

print(f"str():  {str(laptop)}")
print(f"repr(): {repr(laptop)}")
print(f"print(): {laptop}")

# repr() is useful for debugging
products = [Product('Mouse', 29.99), Product('Keyboard', 99.99)]
print(f"\nList representation: {products}")
````

````python
# --- Special Methods: Comparisons and Operations ---

class Vector:
    """A 2D vector with magnitude and operations."""
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __add__(self, other):
        """Add two vectors: v1 + v2."""
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        """Subtract vectors: v1 - v2."""
        return Vector(self.x - other.x, self.y - other.y)
    
    def __mul__(self, scalar):
        """Multiply vector by scalar: v * 2."""
        return Vector(self.x * scalar, self.y * scalar)
    
    def __eq__(self, other):
        """Check equality: v1 == v2."""
        return self.x == other.x and self.y == other.y
    
    def __len__(self):
        """Magnitude: len(v)."""
        return (self.x ** 2 + self.y ** 2) ** 0.5

# Use the special methods
v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(f"v1 = {v1}")
print(f"v2 = {v2}")

print(f"\nv1 + v2 = {v1 + v2}")
print(f"v1 - v2 = {v1 - v2}")
print(f"v1 * 2 = {v1 * 2}")

print(f"\nv1 == v2: {v1 == v2}")
v3 = Vector(3, 4)
print(f"v1 == v3: {v1 == v3}")

print(f"\nMagnitude of v1: {len(v1):.2f}")
print(f"Magnitude of v2: {len(v2):.2f}")
````

## 3. Inheritance — Code Reuse Through Hierarchy

**Inheritance** allows a class to **inherit** attributes and methods from another class. This enables code reuse and hierarchy.

### Basic Hierarchy

```
        Person (parent/base class)
         /          \
    Employee      Student (child/derived classes)
```

### Syntax

```python
class Person:              # Parent class
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def greet(self):
        return f"Hello, I'm {self.name}"

class Employee(Person):    # Child class (inherits from Person)
    def __init__(self, name, age, employee_id, salary):
        super().__init__(name, age)     # Call parent __init__
        self.employee_id = employee_id
        self.salary = salary
    
    def give_raise(self, amount):
        self.salary += amount

emp = Employee('Alice', 30, 'E001', 85000)
print(emp.greet())         # Inherited method: "Hello, I'm Alice"
print(emp.salary)          # Own attribute: 85000
```

### Key Concepts

| Term | Definition |
|------|------------|
| **Parent/Base Class** | Class being inherited from |
| **Child/Derived Class** | Class that inherits |
| **`super()`** | Call parent class methods |
| **Override** | Child redefines a method |

> ⚠️ **Special Case — Always call `super().__init__()`**: When a child class defines `__init__`, you must call the parent's `__init__` to initialize inherited attributes. Otherwise, parent attributes won't be set.

> **Data Science relevance**: Scikit-Learn uses inheritance heavily. `LogisticRegression` inherits from `BaseEstimator`. You inherit from base classes to create custom estimators.

````python
# --- Inheritance Example ---

class Animal:
    """Base class for animals."""
    
    def __init__(self, name, species):
        self.name = name
        self.species = species
    
    def speak(self):
        return f"{self.name} makes a sound"
    
    def __str__(self):
        return f"{self.name} ({self.species})"


class Dog(Animal):  # Dog inherits from Animal
    """Dog is a specific type of Animal."""
    
    def __init__(self, name, breed):
        super().__init__(name, 'Canis familiaris')
        self.breed = breed
    
    def speak(self):  # Override parent method
        return f"{self.name} barks: Woof!"
    
    def fetch(self, item):
        return f"{self.name} fetches the {item}!"


class Cat(Animal):  # Cat inherits from Animal
    def __init__(self, name, color):
        super().__init__(name, 'Felis catus')
        self.color = color
    
    def speak(self):  # Override
        return f"{self.name} meows: Meow!"


# Create objects
dog = Dog('Buddy', 'Golden Retriever')
cat = Cat('Whiskers', 'Orange')

# Inherited methods work
print(f"Dog: {dog}")
print(f"Cat: {cat}")

# Overridden methods
print(f"\n{dog.speak()}")
print(f"{cat.speak()}")

# Own methods
print(f"{dog.fetch('ball')}")

# Check inheritance
print(f"\nIs dog an Animal? {isinstance(dog, Animal)}")
print(f"Is dog a Dog? {isinstance(dog, Dog)}")
print(f"Is cat a Dog? {isinstance(cat, Dog)}")
````

## 4. Encapsulation — Public and Private Attributes

**Encapsulation** is hiding internal details and exposing only what's necessary. Python uses naming conventions (not enforced, but respected).

### Naming Conventions

| Name | Visibility | Use When |
|------|-----------|----------|
| `attribute` | Public | Safe to access/modify directly |
| `_attribute` | Protected | Internal, shouldn't be accessed directly but can be |
| `__attribute` | Private | Strongly suggest no direct access |

### Access Control with Properties

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute
    
    @property
    def balance(self):
        """Getter — read-only access."""
        return self.__balance
    
    def deposit(self, amount):
        """Safe way to modify balance."""
        if amount > 0:
            self.__balance += amount

account = BankAccount(1000)
print(account.balance)     # OK: use property
account.deposit(100)       # OK: use method
account.balance = 999      # Error! (no setter defined)
account.__balance = 999    # Technically possible but DON'T DO THIS
```

### Benefits of Encapsulation

| Benefit | Example |
|---------|----------|
| **Control** | Only allow valid values (age must be > 0) |
| **Consistency** | Can't accidentally break object state |
| **Flexibility** | Can change internal implementation without breaking external code |
| **Debugging** | All changes go through methods (easy to add logging) |

> ⚠️ **Special Case — Python doesn't truly enforce privacy**: `__balance` becomes `_BankAccount__balance` internally (name mangling). You CAN access it if you really want, but DON'T. It's a convention, not a law.

> **Data Science relevance**: Scikit-Learn models use properties. `model.coef_` with trailing underscore means "don't set directly, it's computed."

````python
# --- Encapsulation Example ---

class Person:
    """Person with age validation."""
    
    def __init__(self, name, age):
        self.name = name          # Public
        self._ssn = None           # Protected (shouldn't access directly)
        self.__age = age           # Private
    
    @property
    def age(self):
        """Read-only age property."""
        return self.__age
    
    @age.setter
    def age(self, value):
        """Validate age before setting."""
        if value < 0:
            raise ValueError("Age cannot be negative")
        if value > 150:
            raise ValueError("Age too high (check your data)")
        self.__age = value
    
    def birthday(self):
        """Safe way to increment age."""
        self.__age += 1
    
    def __str__(self):
        return f"{self.name}, age {self.__age}"

# Create and use
alice = Person('Alice', 30)
print(f"Alice: {alice}")
print(f"Age (via property): {alice.age}")

# Using setter with validation
alice.age = 31
print(f"After setting age: {alice}")

# Invalid age rejected
try:
    alice.age = -5
except ValueError as e:
    print(f"Error: {e}")

try:
    alice.age = 200
except ValueError as e:
    print(f"Error: {e}")

# Safe method
alice.birthday()
print(f"After birthday: {alice}")
````

## 5. Polymorphism — Different Behaviors, Same Interface

**Polymorphism** means "many forms." Different classes can have the same method name, but different implementations.

### Example: Shapes

```python
class Shape:
    def area(self):
        pass  # To be overridden

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side
    
    def area(self):
        return self.side ** 2

# Polymorphism in action
shapes = [Circle(5), Square(4), Circle(3)]
for shape in shapes:
    print(f"Area: {shape.area()}")  # Each uses its own area() method!
```

### Benefits

| Benefit | Without Polymorphism | With Polymorphism |
|---------|----------------------|-------------------|
| Code | `if circle: calc_circle_area(c)` `elif square:` ... | `shape.area()` (all same!) |
| Adding shapes | Modify all functions | Add new class, no changes needed |
| Readability | Complex conditionals | Clean, consistent interface |

> **Data Science relevance**: All Scikit-Learn models have the same methods: `.fit()`, `.predict()`, `.score()`. This is polymorphism! You write generic code that works with any model.

````python
# --- Polymorphism Example: ML Models ---

class MLModel:
    """Base class for ML models."""
    
    def fit(self, X, y):
        raise NotImplementedError("Subclass must implement fit()")
    
    def predict(self, X):
        raise NotImplementedError("Subclass must implement predict()")
    
    def score(self, X, y):
        """Default accuracy scorer."""
        predictions = self.predict(X)
        correct = sum(1 for true, pred in zip(y, predictions) if true == pred)
        return correct / len(y)


class SimpleLinearRegression(MLModel):
    """Simple linear regression: y = mx + b."""
    
    def fit(self, X, y):
        # Simplified: just compute slope and intercept
        self.mean_x = sum(X) / len(X)
        self.mean_y = sum(y) / len(y)
        
        num = sum((x - self.mean_x) * (yi - self.mean_y) for x, yi in zip(X, y))
        den = sum((x - self.mean_x) ** 2 for x in X)
        
        self.slope = num / den if den != 0 else 0
        self.intercept = self.mean_y - self.slope * self.mean_x
        return self
    
    def predict(self, X):
        return [self.slope * x + self.intercept for x in X]


class MajorityClassifier(MLModel):
    """Baseline: always predict the most common class."""
    
    def fit(self, X, y):
        # Find most common label
        from collections import Counter
        self.majority = Counter(y).most_common(1)[0][0]
        return self
    
    def predict(self, X):
        return [self.majority for _ in X]


# Use both models with the SAME interface
X_train = [1, 2, 3, 4, 5]
y_train = [2, 4, 5, 4, 5]

X_test = [1.5, 3.5]
y_test = [3, 5]

print("=== Linear Regression ===")
lr = SimpleLinearRegression()
lr.fit(X_train, y_train)
lr_pred = lr.predict(X_test)
print(f"Predictions: {lr_pred}")

print("\n=== Majority Classifier ===")
mc = MajorityClassifier()
mc.fit(X_train, y_train)
mc_pred = mc.predict(X_test)
print(f"Predictions: {mc_pred}")

# Polymorphism: same code, different models
print("\n=== Training Multiple Models ===")
models = [lr, mc]
for i, model in enumerate(models, 1):
    print(f"Model {i}: {model.__class__.__name__}")
    pred = model.predict(X_test)
    print(f"  Prediction: {pred}")
````

## 6. Putting It All Together: Complete Example

A realistic example combining classes, inheritance, encapsulation, and special methods.

````python
# --- Bank Account System ---

class Account:
    """Base class for bank accounts."""
    
    def __init__(self, owner, account_number, initial_balance):
        self.owner = owner
        self.account_number = account_number
        self._balance = initial_balance
        self._transactions = [f"Account opened with ${initial_balance}"]
    
    @property
    def balance(self):
        return self._balance
    
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount
        self._transactions.append(f"Deposit: +${amount}")
        return f"Deposited ${amount}. New balance: ${self._balance}"
    
    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount
        self._transactions.append(f"Withdrawal: -${amount}")
        return f"Withdrew ${amount}. New balance: ${self._balance}"
    
    def __str__(self):
        return f"{self.__class__.__name__}: {self.owner} (#{self.account_number}), Balance: ${self._balance}"
    
    def __repr__(self):
        return f"Account('{self.owner}', '{self.account_number}', {self._balance})"


class SavingsAccount(Account):
    """Savings account with interest."""
    
    def __init__(self, owner, account_number, initial_balance, interest_rate):
        super().__init__(owner, account_number, initial_balance)
        self.interest_rate = interest_rate
    
    def apply_interest(self):
        interest = self._balance * self.interest_rate
        self._balance += interest
        self._transactions.append(f"Interest: +${interest:.2f}")
        return f"Interest applied: ${interest:.2f}"


class CheckingAccount(Account):
    """Checking account with monthly fee."""
    
    def __init__(self, owner, account_number, initial_balance, monthly_fee):
        super().__init__(owner, account_number, initial_balance)
        self.monthly_fee = monthly_fee
    
    def charge_monthly_fee(self):
        if self._balance >= self.monthly_fee:
            self._balance -= self.monthly_fee
            self._transactions.append(f"Monthly fee: -${self.monthly_fee}")
            return f"Monthly fee charged: ${self.monthly_fee}"
        else:
            return "Insufficient funds for monthly fee"


# Use the accounts
savings = SavingsAccount('Alice', 'S001', 1000, 0.05)  # 5% interest
checking = CheckingAccount('Bob', 'C001', 500, 10)     # $10 monthly fee

print("=== Savings Account ===")
print(savings)
print(savings.deposit(200))
print(savings.apply_interest())
print(f"Final balance: ${savings.balance}")

print("\n=== Checking Account ===")
print(checking)
print(checking.deposit(100))
print(checking.withdraw(50))
print(checking.charge_monthly_fee())
print(f"Final balance: ${checking.balance}")
````

| **Classes & Objects** | Classes are blueprints; objects are instances with attributes (data) and methods (behavior) |

### Key Gotchas to Remember
- `self` is always first parameter (but you don't pass it when calling)
- Always call `super().__init__()` in child class if parent has `__init__`
- `__str__` is for users (print friendly), `__repr__` is for developers (debugging)
- Private attributes with `__` are name-mangled but can still be accessed (use as convention, not law)
- `@property` decorator makes methods act like attributes
- Polymorphism works best with consistent interfaces (all subclasses implement same methods)
- Don't create circular inheritance hierarchies

### Data Science Connection
- Scikit-Learn models: `fit()`, `predict()`, `score()` (polymorphism!)
- Custom estimators: inherit from `BaseEstimator`
